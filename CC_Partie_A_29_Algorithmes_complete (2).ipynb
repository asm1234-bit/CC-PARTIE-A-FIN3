## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Régression linéaire (*Linear Regression*, moindres carrés ordinaires) — **famille** : régression supervisée ; **objectif** : prédire une cible numérique $y$ par une combinaison linéaire des variables explicatives.
- **Intuition** : on cherche l'hyperplan qui passe « au plus près » de tous les points, la proximité étant mesurée par la somme des carrés des erreurs verticales (résidus).
- **Étapes** : données d'entrée $(X,y)$ → calcul des coefficients (solution analytique ou descente de gradient) → prédiction $\hat y=X\beta$ pour de nouvelles observations.
- **Critère** : $\displaystyle \hat\beta=\arg\min_{\beta}\sum_{i=1}^{n}\big(y_i-\beta_0-\textstyle\sum_{j=1}^{p}\beta_jx_{ij}\big)^2$, soit $\hat\beta=(X^\top X)^{-1}X^\top y$ ; $n$ = nombre d'observations, $p$ = nombre de variables, $x_{ij}$ = valeur de la variable $j$ pour l'observation $i$, $\beta_0$ = constante, $\beta_j$ = effet d'une unité de $x_j$ sur $y$.
- **Paramètres / hyperparamètres** : les paramètres appris sont $\beta_0,\dots,\beta_p$ ; le modèle de base n'a quasiment aucun hyperparamètre (seulement l'option `fit_intercept`).
- **Avantage 1** : très interprétable (chaque coefficient a une signification économique directe) et très rapide.
- **Avantage 2** : sert de référence solide ; la solution est unique et sans optimisation délicate.
- **Limite 1** : ne capte que des relations linéaires et est sensible aux valeurs extrêmes (carrés des erreurs).
- **Limite 2** : instable quand les variables sont fortement corrélées (multicolinéarité) ou quand $p$ est grand devant $n$.

### Schéma explicatif

**Comment lire ce schéma.** Les points bleus sont les observations ; la droite rouge est le modèle ajusté ; chaque segment gris pointillé est un résidu (écart vertical entre observation et prédiction). L'algorithme choisit la droite qui rend la somme des carrés de ces segments minimale.

```python
def schema():
    g = np.random.default_rng(0); x = np.sort(g.uniform(0, 10, 25)); y = 2 + .8*x + g.normal(0, 1.2, 25)
    b1, b0 = np.polyfit(x, y, 1); yh = b0 + b1*x
    fig, ax = plt.subplots(figsize=(6.5, 4))
    ax.scatter(x, y, c='#2980b9', label=r'observations $(x_i,y_i)$')
    ax.plot(x, yh, 'r-', lw=2, label=r'droite ajustée $\hat y=\beta_0+\beta_1x$')
    ax.vlines(x, y, yh, colors='gray', linestyles=':', label=r'résidus $e_i=y_i-\hat y_i$')
    ax.set_xlabel('variable explicative $x$'); ax.set_ylabel('cible $y$')
    ax.set_title('Régression linéaire : on minimise la somme des carrés des résidus'); ax.legend(); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : estimer la sensibilité du rendement d'un titre à celui du marché (coefficient $\beta$ du MEDAF) à partir de rendements historiques.
- **Cas d'usage 2 (commerce)** : prévoir le chiffre d'affaires d'un point de vente en fonction du prix, du trafic et de la publicité.
- **Données et prérequis** : variables numériques (les catégories sont encodées en indicatrices), cible continue, relation approximativement linéaire, résidus sans structure ; pas besoin de normaliser pour l'estimation mais utile pour comparer les coefficients.
- **Cas peu approprié** : prédire une probabilité de défaut (cible binaire, valeurs hors de [0, 1]) ou une relation fortement non linéaire (saisonnalité, effets de seuil).
- **Métriques** : MAE, RMSE, $R^2$ (idéalement sur un jeu de test) et analyse des résidus.

### Analyse de l'expérience Python

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

**Commentaire.**

**Entrées** : `Xr_ent`, `yr_ent` (210 observations d'entraînement, 5 variables, générées par `make_regression` avec un bruit de 13) ; **hyperparamètres** : aucun (valeurs par défaut). **Sortie** : MAE = **9,35** sur le jeu de test (70 observations).
**Ce que cela permet de conclure** : en moyenne, la prédiction s'écarte de 9,35 unités de la cible. Comme les données sont linéaires par construction avec un bruit gaussien d'écart-type 13, l'erreur absolue moyenne incompressible est d'environ $0{,}8\times13\approx10{,}4$ ; la valeur observée est du même ordre, donc le modèle est proche du plancher de bruit (l'écart provient de la petite taille du test).
**Ce que cela ne permet pas de conclure** : ce résultat ne dit rien sur des données réelles (non linéaires, bruitées autrement) ni sur la stabilité du score (un seul découpage, pas de validation croisée).

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Régression Ridge (*Ridge regression*, régularisation $\ell_2$) — **famille** : régression supervisée ; **objectif** : stabiliser la régression linéaire en pénalisant les coefficients trop grands.
- **Intuition** : on accepte un léger biais pour réduire fortement la variance : les coefficients sont « rétrécis » vers zéro sans jamais être exactement nuls.
- **Étapes** : standardiser les variables → minimiser l'erreur quadratique + pénalité → prédire avec les coefficients rétrécis.
- **Critère** : $\displaystyle \hat\beta=\arg\min_\beta\ \sum_{i=1}^n(y_i-x_i^\top\beta)^2+\alpha\sum_{j=1}^p\beta_j^2$ ; $\alpha\ge0$ = force de la pénalité (si $\alpha=0$ on retrouve les moindres carrés), $x_i$ = vecteur des variables de l'observation $i$.
- **Paramètres / hyperparamètres** : paramètres = $\beta$ ; hyperparamètre principal = $\alpha$ (choisi par validation croisée, ici `alpha=1.0`).
- **Avantage 1** : résout la multicolinéarité et réduit le surapprentissage.
- **Avantage 2** : solution analytique rapide, toujours définie.
- **Limite 1** : ne fait pas de sélection de variables (tous les coefficients restent non nuls).
- **Limite 2** : sensible à l'échelle des variables (standardisation obligatoire) et coefficients biaisés, donc moins directement interprétables.

### Schéma explicatif

**Comment lire ce schéma.** Chaque courbe est un coefficient. À gauche ($\alpha$ petit) on retrouve les moindres carrés ; en allant vers la droite la pénalité augmente et tous les coefficients tendent vers zéro de façon continue. Les deux variables corrélées ($x_1$, $x_2$) se partagent l'effet au lieu de prendre des valeurs opposées extrêmes.

```python
def schema():
    from sklearn.linear_model import Ridge
    g = np.random.default_rng(1); n = 60; z = g.normal(size=n)
    X = np.c_[z + .3*g.normal(size=n), z + .3*g.normal(size=n), g.normal(size=n), g.normal(size=n)]
    y = 3*X[:, 0] + 2*X[:, 1] + .5*X[:, 2] + g.normal(0, 1, n)
    alphas = np.logspace(-2, 3.5, 60)
    coefs = np.array([Ridge(alpha=a).fit(X, y).coef_ for a in alphas])
    fig, ax = plt.subplots(figsize=(6.5, 4))
    for j in range(4):
        ax.plot(alphas, coefs[:, j], label=rf'coefficient $\beta_{j+1}$' + (' (corrélé à $x_2$)' if j == 0 else ' (corrélé à $x_1$)' if j == 1 else ''))
    ax.set_xscale('log'); ax.axhline(0, c='k', lw=.8)
    ax.set_xlabel(r'force de la pénalité $\alpha$ (échelle log)'); ax.set_ylabel('valeur du coefficient')
    ax.set_title('Ridge : les coefficients se rétrécissent vers 0 sans s\'annuler'); ax.legend(fontsize=8); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : modèle multifactoriel de rendement d'actions avec des facteurs corrélés (valeur, taille, momentum), où les coefficients OLS sont instables.
- **Cas d'usage 2 (marketing)** : mesurer l'effet de plusieurs canaux publicitaires corrélés (TV, radio, web) sur les ventes.
- **Données et prérequis** : variables numériques standardisées, cible continue, beaucoup de variables corrélées ; $\alpha$ à régler par validation croisée.
- **Cas peu approprié** : quand on veut identifier un petit nombre de variables utiles (le Lasso est alors préférable) ou quand la relation est non linéaire.
- **Métriques** : MAE, RMSE, $R^2$ sur validation/test ; courbe de validation en fonction de $\alpha$.

### Analyse de l'expérience Python

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

**Commentaire.**

**Entrées** : mêmes données que la régression linéaire, passées dans un `Pipeline` (`StandardScaler` puis `Ridge`) ; **hyperparamètre** : `alpha=1.0`. **Sortie** : MAE = **9,37** (test).
**Lecture** : le score est quasiment identique à celui de la régression linéaire (9,35) : ici les 5 variables sont indépendantes et $n=210\gg p=5$, il n'y a donc presque rien à stabiliser et la pénalité n'apporte ni gain ni perte notable.
**Limites de la conclusion** : un écart de 0,02 est bien inférieur à la variabilité due au découpage ; on ne peut pas déclarer l'un des deux modèles meilleur. L'intérêt de Ridge se verrait avec des variables corrélées ou $p$ proche de $n$, ce que cet exemple ne teste pas.

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Régression Lasso (*Least Absolute Shrinkage and Selection Operator*) — **famille** : régression supervisée parcimonieuse ; **objectif** : prédire une cible numérique tout en sélectionnant automatiquement les variables utiles.
- **Intuition** : la pénalité sur la valeur absolue des coefficients « pousse » les plus faibles exactement à zéro ; il reste un modèle plus simple.
- **Étapes** : standardiser → minimiser erreur quadratique + pénalité $\ell_1$ (descente de coordonnées) → prédire avec les variables retenues.
- **Critère** : $\displaystyle \hat\beta=\arg\min_\beta\ \tfrac{1}{2n}\sum_{i=1}^n(y_i-x_i^\top\beta)^2+\alpha\sum_{j=1}^p|\beta_j|$ ; $\alpha$ = force de pénalité, $|\beta_j|$ = valeur absolue du coefficient $j$.
- **Paramètres / hyperparamètres** : paramètres = $\beta$ (dont beaucoup nuls) ; hyperparamètres = $\alpha$ (ici `alpha=0.1`) et `max_iter` (critère d'arrêt de l'optimisation).
- **Avantage 1** : sélection de variables intégrée, modèle parcimonieux et lisible.
- **Avantage 2** : adapté aux données avec beaucoup de variables ($p$ grand).
- **Limite 1** : parmi des variables très corrélées, il en choisit une de façon arbitraire et instable.
- **Limite 2** : biais des coefficients retenus ; sensible à l'échelle des variables.

### Schéma explicatif

**Comment lire ce schéma.** Les traits pleins sont les trois variables réellement utiles ; les pointillés sont trois variables de pur bruit. Quand $\alpha$ augmente, les variables de bruit tombent à 0 en premier, puis les variables utiles les plus faibles : on lit ainsi l'ordre d'importance des variables.

```python
def schema():
    from sklearn.linear_model import Lasso
    g = np.random.default_rng(2); n = 80
    X = g.normal(size=(n, 6)); y = 4*X[:, 0] - 3*X[:, 1] + 1.0*X[:, 2] + g.normal(0, 1, n)
    alphas = np.logspace(-2, .8, 60)
    coefs = np.array([Lasso(alpha=a, max_iter=10000).fit(X, y).coef_ for a in alphas])
    fig, ax = plt.subplots(figsize=(6.5, 4))
    for j in range(6):
        ax.plot(alphas, coefs[:, j], lw=2 if j < 3 else 1, ls='-' if j < 3 else '--',
                label=rf'$\beta_{j+1}$ ' + ('(variable utile)' if j < 3 else '(bruit)'))
    ax.set_xscale('log'); ax.axhline(0, c='k', lw=.8)
    ax.set_xlabel(r'force de la pénalité $\alpha$ (échelle log)'); ax.set_ylabel('valeur du coefficient')
    ax.set_title('Lasso : les coefficients atteignent exactement 0 (sélection de variables)'); ax.legend(fontsize=7, ncol=2); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : sélectionner, parmi des dizaines d'indicateurs financiers, ceux qui expliquent le score de crédit ou le rendement d'une entreprise.

- **Données et prérequis** : beaucoup de variables dont seulement quelques-unes sont informatives ; standardisation ; $\alpha$ choisi par validation croisée (`LassoCV`).
- **Cas peu approprié** : variables toutes utiles et fortement corrélées (la sélection devient arbitraire), ou relation non linéaire.
- **Métriques** : RMSE/MAE/$R^2$ en validation et test, nombre de coefficients non nuls, stabilité de la sélection sur plusieurs échantillons.

### Analyse de l'expérience Python

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

**Lecture** : le modèle explique environ 98,6 % de la variance de la cible de test, ce qui est cohérent avec des données linéaires peu bruitées. Une pénalité aussi faible n'écarte quasiment aucune variable (les 5 variables sont toutes informatives dans `make_regression` par défaut).

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Régression polynomiale (*Polynomial regression*) — **famille** : régression supervisée (modèle linéaire sur variables transformées) ; **objectif** : capter des relations courbes tout en gardant un modèle linéaire en ses paramètres.
- **Intuition** : on ajoute à $X$ les puissances et produits croisés des variables, puis on ajuste une régression (ici Ridge) sur ces nouvelles variables.
- **Étapes** : standardiser → `PolynomialFeatures` (degré $d$) → régression pénalisée → prédiction.
- **Formule** (une variable) : $\hat y=\beta_0+\beta_1x+\beta_2x^2+\dots+\beta_dx^d$ ; $d$ = degré du polynôme, $\beta_k$ = coefficient du terme de degré $k$ ; avec $p$ variables, le nombre de termes croît comme $\binom{p+d}{d}$.
- **Paramètres / hyperparamètres** : paramètres = les $\beta_k$ ; hyperparamètres = degré $d$ (ici 2) et force de pénalité $\alpha$ (ici 10).
- **Avantage 1** : modélise des courbures et des interactions avec des outils très simples.
- **Avantage 2** : reste interprétable pour de petits degrés.
- **Limite 1** : explosion du nombre de variables et risque de surapprentissage quand $d$ ou $p$ augmentent.
- **Limite 2** : extrapolation hors du domaine d'entraînement très instable (oscillations).

### Schéma explicatif

**Comment lire ce schéma.** Les points sont les observations bruitées d'une sinusoïde (pointillés). Le degré 1 (orange) est trop rigide, le degré 4 (vert) suit bien la forme, le degré 15 (rouge) colle au bruit et oscille entre les points : c'est le surapprentissage que la pénalité Ridge et le choix du degré doivent contrôler.

```python
def schema():
    g = np.random.default_rng(3); x = np.sort(g.uniform(0, 1, 18)); y = np.sin(2*np.pi*x) + g.normal(0, .2, 18); xs = np.linspace(0, 1, 200)
    fig, ax = plt.subplots(figsize=(6.5, 4))
    ax.scatter(x, y, c='k', zorder=3, label='observations')
    ax.plot(xs, np.sin(2*np.pi*xs), 'k--', alpha=.5, label='vraie fonction')
    for d, c, nom in [(1, '#e67e22', 'sous-apprentissage'), (4, '#27ae60', 'bon compromis'), (15, '#c0392b', 'surapprentissage')]:
        p = np.polynomial.Polynomial.fit(x, y, d); ax.plot(xs, p(xs), c=c, lw=2, label=f'degré {d} ({nom})')
    ax.set_ylim(-1.8, 1.8); ax.set_xlabel('$x$'); ax.set_ylabel('$y$')
    ax.set_title('Régression polynomiale : effet du degré'); ax.legend(fontsize=8); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : approximer la relation non linéaire entre la maturité d'un titre et son rendement (courbe des taux).
- **Cas d'usage 2 (logistique)** : modéliser le coût de transport en fonction de la distance avec des rendements décroissants.
- **Données et prérequis** : peu de variables numériques standardisées, effet courbe visible sur un nuage de points, degré faible.
- **Cas peu approprié** : beaucoup de variables (explosion combinatoire) ou besoin d'extrapoler au-delà des valeurs observées.
- **Métriques** : MAE, RMSE, $R^2$ en validation croisée en fonction du degré ; erreur de validation vs erreur d'entraînement.

### Analyse de l'expérience Python

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

**Commentaire.**

**Entrées** : mêmes données ; pipeline `StandardScaler` → `PolynomialFeatures(degree=2)` → `Ridge(alpha=10)` ; **sortie** : MAE = **10,90** (test).
**Lecture** : c'est moins bon que la régression linéaire simple (9,35). Les données ont été générées par une relation exactement linéaire : les 15 termes d'ordre 2 ajoutés ne contiennent que du bruit et dégradent légèrement la prédiction malgré la pénalité.
**Limites** : ce résultat est propre aux données synthétiques linéaires ; sur des données réelles courbes, la conclusion peut s'inverser. Le degré n'a pas été optimisé (pas de validation), donc on ne peut pas affirmer que le degré 2 est mauvais en général.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Arbre de régression (*Regression tree*, CART) — **famille** : régression supervisée non paramétrique ; **objectif** : prédire une valeur numérique par une suite de règles « si variable ≤ seuil ».
- **Intuition** : on découpe l'espace des variables en rectangles ; dans chaque rectangle, la prédiction est la moyenne des cibles d'entraînement.
- **Étapes** : choisir la meilleure coupure (variable, seuil) → répéter récursivement dans chaque sous-groupe → s'arrêter (profondeur, taille minimale) → prédire la moyenne de la feuille.
- **Critère** : à chaque nœud on minimise $\displaystyle \sum_{i\in G}(y_i-\bar y_G)^2+\sum_{i\in D}(y_i-\bar y_D)^2$ ; $G$ et $D$ = groupes gauche/droite créés par la coupure, $\bar y_G$, $\bar y_D$ = moyennes des cibles dans chaque groupe.
- **Paramètres / hyperparamètres** : paramètres = variables et seuils de coupure, valeurs des feuilles ; hyperparamètres = `max_depth` (ici 5), `min_samples_leaf`, etc.
- **Avantage 1** : lisible (règles « si… alors ») et sans standardisation.
- **Avantage 2** : capte non-linéarités et interactions automatiquement.
- **Limite 1** : très instable (un petit changement des données change l'arbre) et sujet au surapprentissage.
- **Limite 2** : prédiction constante par morceaux : mauvaise approximation des relations linéaires et pas d'extrapolation.

### Schéma explicatif

**Comment lire ce schéma.** La courbe en escaliers est la prédiction : chaque palier est la moyenne des points d'une feuille. Les lignes verticales pointillées sont les seuils de coupure choisis par l'arbre de profondeur 2. En augmentant la profondeur (rouge) l'escalier devient plus fin et suit davantage le bruit.

```python
def schema():
    from sklearn.tree import DecisionTreeRegressor
    g = np.random.default_rng(4); x = np.sort(g.uniform(0, 10, 80)); y = np.sin(x) + .3*x + g.normal(0, .25, 80); xs = np.linspace(0, 10, 400)[:, None]
    t2 = DecisionTreeRegressor(max_depth=2).fit(x[:, None], y); t5 = DecisionTreeRegressor(max_depth=5).fit(x[:, None], y)
    fig, ax = plt.subplots(figsize=(6.8, 4))
    ax.scatter(x, y, s=16, c='gray', label='observations')
    ax.plot(xs, t2.predict(xs), c='#2980b9', lw=2, label='profondeur 2 (4 feuilles)')
    ax.plot(xs, t5.predict(xs), c='#c0392b', lw=1.5, label='profondeur 5 (jusqu\'à 32 feuilles)')
    for s in sorted(t2.tree_.threshold[t2.tree_.feature >= 0]): ax.axvline(s, c='#2980b9', ls=':', lw=1)
    ax.set_xlabel('variable $x$'); ax.set_ylabel('cible $y$'); ax.set_title('Arbre de régression : prédiction en escaliers'); ax.legend(fontsize=8); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : estimer la perte attendue d'un prêt selon des règles simples (revenu, ancienneté, montant) facilement expliquables au client.
- **Cas d'usage 2 (logistique)** : prévoir un délai de livraison par segments (distance, type de produit, saison).
- **Données et prérequis** : variables numériques ou catégorielles, relations non linéaires ou à seuils ; réglage de la profondeur.
- **Cas peu approprié** : relation lisse et linéaire avec peu de données (le modèle linéaire est meilleur), ou extrapolation hors de la plage observée.
- **Métriques** : MAE, RMSE, $R^2$ ; profondeur/nombre de feuilles pour juger la complexité.

### Analyse de l'expérience Python

```python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

**Commentaire.**

**Entrées** : mêmes données ; **hyperparamètres** : `max_depth=5`, `random_state=42`. **Sortie** : MAE = **37,72** (test).
**Lecture** : c'est de loin la plus mauvaise MAE du bloc régression (régression linéaire : 9,35). La cible est une fonction linéaire de 5 variables : un arbre de profondeur 5 (au plus 32 feuilles pour 210 points) ne peut la représenter que par paliers, d'où une erreur environ quatre fois plus grande.
**Limites** : ce n'est pas une preuve que les arbres sont « mauvais » : sur des relations à seuils ou avec interactions, l'ordre s'inverse souvent. La profondeur n'a pas été optimisée.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Forêt aléatoire pour la régression (*Random Forest Regressor*) — **famille** : régression supervisée (ensemble de bagging) ; **objectif** : réduire l'instabilité d'un arbre en moyennant de nombreux arbres.
- **Intuition** : chaque arbre voit un échantillon bootstrap différent et, à chaque coupure, un sous-ensemble aléatoire de variables ; leurs erreurs se compensent en moyenne.
- **Étapes** : tirer $B$ échantillons bootstrap → entraîner un arbre profond sur chacun → prédire en moyennant les $B$ prédictions.
- **Formule** : $\displaystyle \hat f(x)=\frac1B\sum_{b=1}^{B}T_b(x)$ ; $B$ = nombre d'arbres, $T_b$ = $b$-ième arbre, $x$ = observation à prédire.
- **Paramètres / hyperparamètres** : paramètres = structure des arbres ; hyperparamètres = `n_estimators` ($B$=80 ici), `max_depth`, `max_features`, `min_samples_leaf`.
- **Avantage 1** : robuste, peu sensible au réglage, bonne performance par défaut sur données tabulaires.
- **Avantage 2** : fournit une importance des variables et une estimation d'erreur *out-of-bag*.
- **Limite 1** : moins interprétable qu'un arbre unique et plus coûteuse en mémoire et en temps.
- **Limite 2** : ne peut pas extrapoler (prédictions bornées par les cibles observées) et reste en escaliers sur les relations linéaires.

### Schéma explicatif

**Comment lire ce schéma.** Chaque courbe bleu clair est un arbre entraîné sur un échantillon bootstrap différent : ils sont irréguliers et divergent. La courbe rouge épaisse est leur moyenne : elle est nettement plus lisse et plus stable, c'est le gain de variance du bagging.

```python
def schema():
    from sklearn.ensemble import RandomForestRegressor
    g = np.random.default_rng(5); x = np.sort(g.uniform(0, 10, 100)); y = np.sin(x) + .3*x + g.normal(0, .3, 100); xs = np.linspace(0, 10, 400)[:, None]
    rf = RandomForestRegressor(n_estimators=40, max_depth=6, random_state=0).fit(x[:, None], y)
    fig, ax = plt.subplots(figsize=(6.8, 4))
    ax.scatter(x, y, s=14, c='gray', label='observations')
    for k, t in enumerate(rf.estimators_[:12]):
        ax.plot(xs, t.predict(xs), c='#85c1e9', lw=.8, alpha=.8, label='12 arbres individuels' if k == 0 else None)
    ax.plot(xs, rf.predict(xs), c='#c0392b', lw=2.5, label='moyenne de la forêt (40 arbres)')
    ax.set_xlabel('variable $x$'); ax.set_ylabel('cible $y$'); ax.set_title('Forêt aléatoire : la moyenne lisse les arbres instables'); ax.legend(fontsize=8); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : estimer la valeur d'un bien immobilier ou le montant de recouvrement attendu avec de nombreux facteurs hétérogènes.
- **Cas d'usage 2 (commerce)** : prévoir la demande d'un produit selon prix, promotions, saison et canal.
- **Données et prérequis** : tabulaires, mixtes (numériques/catégorielles encodées), assez d'observations ; pas de standardisation nécessaire.
- **Cas peu approprié** : séries temporelles avec tendance à extrapoler, ou contexte exigeant des règles transparentes.
- **Métriques** : MAE, RMSE, $R^2$, erreur *out-of-bag*, importance des variables.

### Analyse de l'expérience Python

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

**Commentaire.**

**Entrées** : mêmes données ; **hyperparamètres** : `n_estimators=80`, `random_state=42` (autres valeurs par défaut). **Sortie** : MAE = **24,37** (test).
**Lecture** : bien meilleur qu'un arbre unique de profondeur 5 (37,72) : la moyenne de 80 arbres réduit fortement la variance. Mais le résultat reste plus de deux fois moins bon que la régression linéaire (9,35) car la vraie relation est linéaire et la forêt l'approche par paliers.
**Limites** : on ne sait pas si un réglage (profondeur, `max_features`) réduirait l'écart, et un seul découpage ne mesure pas la variabilité.

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Gradient Boosting pour la régression (*Gradient Boosting Regressor*, ici version à histogrammes `HistGradientBoostingRegressor`) — **famille** : régression supervisée (ensemble séquentiel) ; **objectif** : construire un modèle précis en corrigeant progressivement les erreurs.
- **Intuition** : on part d'une prédiction constante, puis chaque nouvel petit arbre apprend à prédire les erreurs (résidus) du modèle courant.
- **Étapes** : initialiser avec la moyenne → pour $m=1..M$ : calculer les résidus, ajuster un arbre dessus, l'ajouter pondéré par le pas $\nu$ → prédiction finale = somme.
- **Formule** : $\displaystyle F_m(x)=F_{m-1}(x)+\nu\,h_m(x)$ où $h_m$ est l'arbre ajusté sur les résidus $r_i=y_i-F_{m-1}(x_i)$ ; $\nu$ = taux d'apprentissage, $M$ = nombre d'itérations.
- **Paramètres / hyperparamètres** : paramètres = arbres $h_m$ ; hyperparamètres = `max_iter` ($M$=80), `learning_rate` ($\nu$), `max_depth`, `l2_regularization`.
- **Avantage 1** : très bonne précision sur données tabulaires, gère les non-linéarités et interactions.
- **Avantage 2** : la version à histogrammes est rapide, gère nativement les valeurs manquantes.
- **Limite 1** : sensible au réglage ($\nu$, $M$, profondeur) et au surapprentissage si $M$ est trop grand.
- **Limite 2** : peu interprétable et entraînement séquentiel (difficile à paralléliser par arbre).

### Schéma explicatif

**Comment lire ce schéma.** Chaque courbe est la prédiction du modèle après un certain nombre d'arbres : avec 1 arbre (orange) on voit une forme très grossière ; avec 5 (vert) la tendance apparaît ; avec 60 (rouge) la courbe épouse les données. Chaque arbre ajouté corrige les erreurs restantes du modèle précédent.

```python
def schema():
    from sklearn.ensemble import GradientBoostingRegressor
    g = np.random.default_rng(6); x = np.sort(g.uniform(0, 10, 100)); y = np.sin(x) + .3*x + g.normal(0, .25, 100); xs = np.linspace(0, 10, 400)[:, None]
    gb = GradientBoostingRegressor(n_estimators=60, max_depth=2, learning_rate=.3, random_state=0).fit(x[:, None], y)
    etapes = {m: p for m, p in enumerate(gb.staged_predict(xs), 1)}
    fig, ax = plt.subplots(figsize=(6.8, 4))
    ax.scatter(x, y, s=14, c='gray', label='observations')
    for m, c in [(1, '#f5b041'), (5, '#52be80'), (60, '#c0392b')]:
        ax.plot(xs, etapes[m], c=c, lw=2, label=f'après {m} arbre(s)')
    ax.set_xlabel('variable $x$'); ax.set_ylabel('cible $y$'); ax.set_title('Gradient boosting : le modèle s\'affine à chaque arbre ajouté'); ax.legend(fontsize=8); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : prévoir la volatilité ou la perte en cas de défaut à partir de nombreux indicateurs.
- **Cas d'usage 2 (marketing)** : prédire la valeur vie client (CLV) à partir de l'historique d'achats.
- **Données et prérequis** : tabulaires de taille moyenne à grande ; réglage par validation (arrêt précoce conseillé).
- **Cas peu approprié** : très peu de données ou besoin d'un modèle explicable par construction ; relations purement linéaires (un modèle linéaire suffit).
- **Métriques** : MAE, RMSE, $R^2$, courbe d'erreur de validation selon le nombre d'itérations.

### Analyse de l'expérience Python

```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele=HistGradientBoostingRegressor(max_iter=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

**Commentaire.**

**Entrées** : mêmes données ; **hyperparamètres** : `max_iter=80`, `random_state=42` (pas d'arrêt précoce, pas de réglage). **Sortie** : MAE = **23,27** (test).
**Lecture** : légèrement meilleur que la forêt aléatoire (24,37) mais très loin de la régression linéaire (9,35) : même cause (cible linéaire).
**Limites** : l'écart de 1,1 point entre boosting et forêt est faible face à la variabilité d'un test de 70 observations ; on ne peut pas classer ces deux méthodes sur cette seule exécution.

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Régression logistique (*Logistic Regression*) — **famille** : classification supervisée ; **objectif** : estimer la probabilité d'appartenir à une classe (binaire ici).
- **Intuition** : on calcule un score linéaire puis on le transforme par la fonction sigmoïde pour obtenir une probabilité entre 0 et 1.
- **Étapes** : standardiser → ajuster les coefficients par maximum de vraisemblance → calculer $P(y=1|x)$ → classer selon un seuil (0,5 par défaut).
- **Formule** : $\displaystyle P(y=1\mid x)=\sigma(\beta_0+x^\top\beta)=\frac{1}{1+e^{-(\beta_0+x^\top\beta)}}$ ; on minimise la log-perte $-\sum_i\big[y_i\ln p_i+(1-y_i)\ln(1-p_i)\big]$ ; $p_i$ = probabilité prédite pour l'observation $i$.
- **Paramètres / hyperparamètres** : paramètres = $\beta_0,\beta$ ; hyperparamètres = force de régularisation `C` (inverse de la pénalité, défaut 1), type de pénalité, seuil de décision.
- **Avantage 1** : sortie probabiliste calibrée et coefficients interprétables (rapports de cotes $e^{\beta_j}$).
- **Avantage 2** : rapide, robuste, excellente référence en scoring.
- **Limite 1** : frontière de décision linéaire (ne capte pas les interactions sans variables ajoutées).
- **Limite 2** : sensible aux variables corrélées et aux valeurs extrêmes ; performance dégradée sur classes déséquilibrées sans ajustement du seuil.

### Schéma explicatif

**Comment lire ce schéma.** Les points rouges/bleus sont les observations de la classe 1/0 (placées à $y=1$ ou $y=0$). La courbe noire en S est la probabilité prédite : elle croise le seuil 0,5 à la frontière de décision (pointillé rouge). Plus on s'éloigne de cette frontière, plus la prédiction est « sûre ».

```python
def schema():
    g = np.random.default_rng(7); x = np.r_[g.normal(-1.2, 1, 40), g.normal(1.2, 1, 40)]; y = np.r_[np.zeros(40), np.ones(40)]
    from sklearn.linear_model import LogisticRegression
    m = LogisticRegression().fit(x[:, None], y); xs = np.linspace(-5, 5, 300)
    fig, ax = plt.subplots(figsize=(6.5, 4))
    ax.scatter(x, y, c=y, cmap='coolwarm', edgecolor='k', s=26, label='observations (classe 0 ou 1)')
    ax.plot(xs, m.predict_proba(xs[:, None])[:, 1], 'k-', lw=2, label=r'probabilité $P(y=1\mid x)$ (sigmoïde)')
    ax.axhline(.5, c='gray', ls='--', label='seuil 0,5'); ax.axvline(-m.intercept_[0]/m.coef_[0, 0], c='#c0392b', ls=':', label='frontière de décision')
    ax.set_xlabel('score / variable $x$'); ax.set_ylabel('probabilité'); ax.set_title('Régression logistique : de la variable à la probabilité'); ax.legend(fontsize=8); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : scoring de crédit — probabilité de défaut d'un emprunteur.
- **Cas d'usage 2 (marketing)** : probabilité qu'un client réponde à une campagne ou résilie son abonnement (*churn*).
- **Données et prérequis** : cible binaire, variables numériques standardisées (catégories encodées), échantillon suffisant par classe.
- **Cas peu approprié** : frontière très non linéaire (images, signaux), ou besoin d'une très haute précision sur des interactions complexes.
- **Métriques** : F1, précision, rappel, ROC-AUC, PR-AUC (si classes déséquilibrées), calibration (courbe de fiabilité).

### Analyse de l'expérience Python

```python
from sklearn.linear_model import LogisticRegression
modele=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

**Commentaire.**

**Entrées** : `Xc_ent`/`yc_ent` (300 observations, 6 variables, générées par `make_classification`) ; pipeline `StandardScaler` + `LogisticRegression(max_iter=1000)`. **Sortie** : F1 = **0,811** (test de 100 observations).
**Lecture** : le F1 (moyenne harmonique de la précision et du rappel de la classe positive) vaut 0,81 sur 1. C'est inférieur aux modèles non linéaires du bloc (boosting 0,950, forêt 0,918), ce qui suggère que la frontière de classes générée n'est pas parfaitement linéaire.
**Limites** : sur 100 observations, une ou deux erreurs de plus font varier le F1 de 1 à 2 points ; la précision et le rappel ne sont pas affichés séparément, donc on ne sait pas quel type d'erreur domine.

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : k plus proches voisins (*k-Nearest Neighbors*, KNN) — **famille** : classification supervisée (apprentissage « paresseux » à base d'exemples) ; **objectif** : classer une observation selon la classe majoritaire de ses voisins.
- **Intuition** : « dis-moi qui sont tes voisins, je te dirai qui tu es » : une observation ressemble à celles qui lui sont proches.
- **Étapes** : mémoriser les données d'entraînement → pour une nouvelle observation, calculer les distances à toutes les observations → retenir les $k$ plus proches → voter.
- **Formule** : $\hat y(x)=\arg\max_c\sum_{i\in\mathcal N_k(x)}\mathbf 1[y_i=c]$ avec la distance euclidienne $d(x,x_i)=\sqrt{\sum_j(x_j-x_{ij})^2}$ ; $\mathcal N_k(x)$ = ensemble des $k$ voisins les plus proches de $x$, $c$ = classe candidate.
- **Paramètres / hyperparamètres** : pas de paramètres appris (les données sont le modèle) ; hyperparamètres = $k$ (ici 5), métrique de distance, pondération des voisins.
- **Avantage 1** : très simple, aucune hypothèse sur la forme de la frontière.
- **Avantage 2** : s'adapte naturellement à des frontières complexes.
- **Limite 1** : coûteux à la prédiction sur de gros jeux de données et sensible à l'échelle des variables (standardisation indispensable).
- **Limite 2** : souffre de la malédiction de la dimension et du bruit sur les variables inutiles.

### Schéma explicatif

**Comment lire ce schéma.** L'étoile jaune est une nouvelle observation à classer. Le cercle pointillé délimite ses 5 plus proches voisins (reliés par des traits). La classe prédite est celle qui est majoritaire parmi eux ; le titre indique le décompte des votes.

```python
def schema():
    from sklearn.neighbors import NearestNeighbors
    g = np.random.default_rng(8); X = np.r_[g.normal([-1, -1], .9, (30, 2)), g.normal([1.3, 1.2], .9, (30, 2))]; y = np.r_[np.zeros(30), np.ones(30)]
    q = np.array([[.3, .2]]); d, idx = NearestNeighbors(n_neighbors=5).fit(X).kneighbors(q)
    fig, ax = plt.subplots(figsize=(5.8, 5))
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolor='k', s=26, label='observations connues (2 classes)')
    for i in idx[0]: ax.plot([q[0, 0], X[i, 0]], [q[0, 1], X[i, 1]], 'k-', lw=.8)
    ax.add_patch(Circle(q[0], d[0, -1], fill=False, ec='k', ls='--')); ax.scatter(*q[0], marker='*', s=260, c='gold', edgecolor='k', zorder=5, label='nouvelle observation')
    votes = y[idx[0]].astype(int); ax.set_title(f'KNN (k=5) : votes {np.bincount(votes, minlength=2)[0]} vs {np.bincount(votes, minlength=2)[1]} → classe {np.bincount(votes).argmax()}')
    ax.set_xlabel('$x_1$'); ax.set_ylabel('$x_2$'); ax.legend(fontsize=8, loc='lower right'); ax.set_aspect('equal'); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (commerce)** : système de recommandation simple : trouver les clients les plus semblables pour suggérer des produits.
- **Cas d'usage 2 (finance)** : détecter un profil de transaction similaire à des fraudes déjà étiquetées.
- **Données et prérequis** : variables numériques standardisées, dimension modérée, jeu d'entraînement représentatif et propre.
- **Cas peu approprié** : très grand volume (prédiction lente), très grande dimension ou nombreuses variables non pertinentes.
- **Métriques** : F1, précision, rappel, ROC-AUC (via les fréquences de vote), choix de $k$ par validation croisée.

### Analyse de l'expérience Python

```python
from sklearn.neighbors import KNeighborsClassifier
modele=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

**Commentaire.**

**Entrées** : mêmes données de classification, pipeline `StandardScaler` + `KNeighborsClassifier` ; **hyperparamètre** : `n_neighbors=5`. **Sortie** : F1 = **0,889**.
**Lecture** : meilleur que la régression logistique (0,811) et que Naive Bayes (0,755) : la frontière de classes n'est pas linéaire et les voisins la capturent. On est néanmoins un peu en dessous de la forêt aléatoire (0,918) et du boosting (0,950).
**Limites** : $k$ n'a pas été réglé ; 100 observations de test rendent les écarts de quelques points peu significatifs.

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Classifieur bayésien naïf (*Gaussian Naive Bayes*) — **famille** : classification supervisée probabiliste ; **objectif** : classer à partir de la règle de Bayes en supposant les variables indépendantes entre elles dans chaque classe.
- **Intuition** : pour chaque classe, on apprend la distribution (gaussienne) de chaque variable ; on choisit la classe la plus plausible compte tenu des valeurs observées.
- **Étapes** : estimer pour chaque classe les probabilités a priori, moyennes et variances → pour une observation, multiplier les vraisemblances → prendre la classe de plus forte probabilité a posteriori.
- **Formule** : $\displaystyle \hat y=\arg\max_c\ P(c)\prod_{j=1}^{p}\mathcal N(x_j;\mu_{cj},\sigma_{cj}^2)$ ; $P(c)$ = proportion de la classe $c$, $\mu_{cj},\sigma_{cj}^2$ = moyenne et variance de la variable $j$ dans la classe $c$.
- **Paramètres / hyperparamètres** : paramètres = $P(c),\mu_{cj},\sigma^2_{cj}$ ; hyperparamètre principal : `var_smoothing` (lissage des variances).
- **Avantage 1** : extrêmement rapide, fonctionne avec très peu de données.
- **Avantage 2** : naturellement probabiliste et robuste en grande dimension.
- **Limite 1** : l'hypothèse d'indépendance est rarement vraie (variables corrélées) et biaise les probabilités.
- **Limite 2** : l'hypothèse gaussienne est inadaptée aux variables asymétriques ou discrètes sans transformation.

### Schéma explicatif

**Comment lire ce schéma.** Chaque courbe représente, pour une classe, la densité gaussienne de la variable pondérée par la fréquence de la classe. Pour une valeur $x$ donnée (trait vertical), on compare la hauteur des deux courbes : la classe de la courbe la plus haute est prédite ; le titre donne la probabilité a posteriori correspondante. Avec plusieurs variables, ces hauteurs sont multipliées.

```python
def schema():
    from scipy.stats import norm
    xs = np.linspace(-6, 7, 400); p0, p1 = .6, .4
    f0 = p0*norm.pdf(xs, -1, 1.2); f1 = p1*norm.pdf(xs, 2, 1.5); x0 = .6
    fig, ax = plt.subplots(figsize=(6.8, 4))
    ax.plot(xs, f0, c='#2980b9', lw=2, label=r'$P(c=0)\,\mathcal{N}(x;\mu_0,\sigma_0^2)$')
    ax.plot(xs, f1, c='#c0392b', lw=2, label=r'$P(c=1)\,\mathcal{N}(x;\mu_1,\sigma_1^2)$')
    ax.axvline(x0, c='k', ls='--'); ax.scatter([x0, x0], [p0*norm.pdf(x0, -1, 1.2), p1*norm.pdf(x0, 2, 1.5)], c=['#2980b9', '#c0392b'], zorder=5)
    post1 = f1[np.abs(xs - x0).argmin()]/(f0[np.abs(xs - x0).argmin()] + f1[np.abs(xs - x0).argmin()])
    ax.set_title(f'Naive Bayes (1 variable) : pour x={x0}, P(classe 1 | x) ≈ {post1:.2f}'); ax.set_xlabel('valeur de la variable $x$'); ax.set_ylabel('vraisemblance × probabilité a priori'); ax.legend(fontsize=8); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance / texte)** : classer des actualités ou communiqués financiers (positif/négatif) à partir de la fréquence des mots (variante multinomiale).
- **Cas d'usage 2 (gestion)** : filtrage de courriels indésirables ou tri automatique de demandes clients.
- **Données et prérequis** : variables peu corrélées, modèle gaussien adapté à des variables continues approximativement normales ; peu de données suffisent.
- **Cas peu approprié** : variables fortement corrélées (revenus et patrimoine, par exemple) où les probabilités deviennent mal calibrées.
- **Métriques** : F1, rappel, précision, ROC-AUC, log-perte (pour juger la calibration).

### Analyse de l'expérience Python

```python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

**Commentaire.**

**Entrées** : mêmes données ; **hyperparamètres** : valeurs par défaut (`GaussianNB()`). **Sortie** : F1 = **0,755**.
**Lecture** : c'est le plus faible score de la classification sur cet exemple. `make_classification` crée des variables informatives *combinées* (non indépendantes dans chaque classe) : l'hypothèse naïve d'indépendance est donc violée, ce qui explique en partie le résultat.
**Limites** : l'explication par la violation de l'indépendance est une hypothèse plausible mais non testée ici (il faudrait examiner la matrice de corrélation intra-classe). Aucun réglage n'a été fait.

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Arbre de décision pour la classification (*Decision Tree Classifier*, CART) — **famille** : classification supervisée ; **objectif** : prédire une classe par une suite de questions simples sur les variables.
- **Intuition** : chaque question « variable ≤ seuil ? » sépare les observations en groupes de plus en plus purs jusqu'à des feuilles quasi homogènes.
- **Étapes** : chercher la coupure qui réduit le plus l'impureté → répéter récursivement → s'arrêter (profondeur, taille) → la feuille prédit la classe majoritaire.
- **Critère** : impureté de Gini $G=\sum_{c}p_c(1-p_c)=1-\sum_cp_c^2$ ; $p_c$ = proportion de la classe $c$ dans le nœud ; la coupure choisie maximise la baisse de $G$ pondérée par la taille des enfants.
- **Paramètres / hyperparamètres** : paramètres = variables et seuils de coupure, classes des feuilles ; hyperparamètres = `max_depth` (ici 4), `min_samples_leaf`, `criterion`.
- **Avantage 1** : règles lisibles, facilement expliquées à un non-spécialiste.
- **Avantage 2** : pas de standardisation, gère variables mixtes et interactions.
- **Limite 1** : forte variance (instable) et tendance à surapprendre si profond.
- **Limite 2** : frontières en rectangles parallèles aux axes (mauvaise pour les frontières obliques).

### Schéma explicatif

**Comment lire ce schéma.** À gauche : l'espace $(x_1,x_2)$ est découpé en rectangles par des frontières parallèles aux axes ; chaque rectangle est coloré selon la classe prédite. À droite : l'arbre correspondant, à lire de haut en bas — chaque nœud pose une question (si vrai, on va à gauche), indique son impureté Gini, le nombre d'observations et la classe majoritaire ; les feuilles (en bas) donnent la décision.

```python
def schema():
    from sklearn.tree import DecisionTreeClassifier, plot_tree
    g = np.random.default_rng(9); X = np.r_[g.normal([-1, -1], 1, (40, 2)), g.normal([1.4, 1.2], 1, (40, 2))]; y = np.r_[np.zeros(40), np.ones(40)].astype(int)
    t = DecisionTreeClassifier(max_depth=2, random_state=0).fit(X, y)
    fig, ax = plt.subplots(1, 2, figsize=(11, 4.2), gridspec_kw={'width_ratios': [1, 1.3]})
    frontiere(ax[0], t, X, y, 'Régions de décision (rectangles)')
    plot_tree(t, feature_names=['$x_1$', '$x_2$'], class_names=['classe 0', 'classe 1'], filled=True, impurity=True, ax=ax[1], fontsize=7)
    ax[1].set_title('Arbre de profondeur 2 : questions successives'); plt.tight_layout(); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : décision d'octroi de crédit par règles explicites exigées par la réglementation.
- **Cas d'usage 2 (logistique)** : classer des colis (prioritaire/standard) selon poids, destination, délai.
- **Données et prérequis** : tabulaires, mixtes ; limiter la profondeur ou élaguer pour éviter le surapprentissage.
- **Cas peu approprié** : frontière oblique ou lisse, petits échantillons bruités (arbre instable).
- **Métriques** : F1, précision, rappel, ROC-AUC ; profondeur et nombre de feuilles pour la complexité.

### Analyse de l'expérience Python

```python
from sklearn.tree import DecisionTreeClassifier
modele=DecisionTreeClassifier(max_depth=4,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

**Commentaire.**

**Entrées** : mêmes données ; **hyperparamètres** : `max_depth=4`, `random_state=42`. **Sortie** : F1 = **0,891**.
**Lecture** : très proche de KNN (0,889) et nettement supérieur à la régression logistique (0,811) et au Naive Bayes (0,755) ; un arbre de profondeur 4 suffit donc à approcher une frontière non linéaire. Il reste en dessous de la forêt (0,918) et du boosting (0,950), ce qui est attendu : ils réduisent la variance d'un arbre seul.
**Limites** : pas de réglage de la profondeur ; la cellule n'affiche pas la structure de l'arbre, donc on ne peut pas dire quelles variables sont utilisées.

## 12. Forêt aléatoire — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Forêt aléatoire pour la classification (*Random Forest Classifier*) — **famille** : classification supervisée (ensemble de bagging) ; **objectif** : stabiliser les arbres de décision en les faisant voter.
- **Intuition** : chaque arbre est entraîné sur un échantillon bootstrap et choisit ses coupures parmi un sous-ensemble aléatoire de variables ; leurs erreurs, peu corrélées, se compensent.
- **Étapes** : tirer $B$ échantillons bootstrap → entraîner un arbre sur chacun → classer par vote majoritaire (ou moyenne des probabilités).
- **Formule** : $\displaystyle \hat y(x)=\arg\max_c\sum_{b=1}^{B}\mathbf 1\big[T_b(x)=c\big]$ ; $B$ = nombre d'arbres, $T_b(x)$ = classe prédite par l'arbre $b$, $c$ = classe candidate.
- **Paramètres / hyperparamètres** : paramètres = structure des arbres ; hyperparamètres = `n_estimators` (80 ici), `max_depth`, `max_features`, `min_samples_leaf`, `class_weight`.
- **Avantage 1** : performante et robuste avec peu de réglage, gère variables mixtes et non-linéarités.
- **Avantage 2** : donne l'importance des variables et une erreur *out-of-bag* sans jeu de validation séparé.
- **Limite 1** : modèle peu interprétable (boîte grise) et plus lourd qu'un arbre unique.
- **Limite 2** : probabilités parfois mal calibrées et performance limitée sur classes très déséquilibrées sans pondération.

### Schéma explicatif

**Comment lire ce schéma.** Les trois premiers graphiques montrent les régions de décision de trois arbres du même ensemble : chacun est irrégulier et différent. Le dernier graphique montre le vote de la forêt : la frontière est plus régulière car les erreurs individuelles se compensent.

```python
def schema():
    from sklearn.ensemble import RandomForestClassifier
    g = np.random.default_rng(10); X = np.r_[g.normal([-1, -1], 1.1, (50, 2)), g.normal([1.2, 1.2], 1.1, (50, 2))]; y = np.r_[np.zeros(50), np.ones(50)].astype(int)
    rf = RandomForestClassifier(n_estimators=30, max_depth=3, random_state=0).fit(X, y)
    fig, ax = plt.subplots(1, 4, figsize=(15, 3.6))
    for k in range(3): frontiere(ax[k], rf.estimators_[k], X, y, f'arbre n°{k+1} (échantillon bootstrap)')
    frontiere(ax[3], rf, X, y, 'vote de la forêt (30 arbres)'); plt.tight_layout(); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : scoring de défaut de crédit ou détection de fraude sur des données tabulaires hétérogènes.
- **Cas d'usage 2 (marketing)** : prédire le départ d'un client (*churn*) à partir de son comportement.
- **Données et prérequis** : tabulaires, mixtes ; pas de standardisation ; gérer le déséquilibre (`class_weight`).
- **Cas peu approprié** : données non structurées (images, texte brut) ou contrainte d'explicabilité stricte.
- **Métriques** : F1, rappel, précision, ROC-AUC, PR-AUC ; importance des variables.

### Analyse de l'expérience Python

```python
from sklearn.ensemble import RandomForestClassifier
modele=RandomForestClassifier(n_estimators=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

**Commentaire.**

**Entrées** : `Xc_ent`/`yc_ent` ; **hyperparamètres** : `n_estimators=80`, `random_state=42`. **Sortie** : F1 = **0,918**.
**Lecture** : amélioration nette par rapport à l'arbre unique de profondeur 4 (0,891), cohérente avec la réduction de variance du bagging. C'est le deuxième meilleur score du bloc derrière le boosting (0,950).
**Limites** : 100 observations de test ; aucune mesure de variabilité, donc l'écart de 3 points avec le boosting n'est pas démontré significatif.

## 13. SVM

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Machine à vecteurs de support (*Support Vector Machine*, SVM) — **famille** : classification supervisée ; **objectif** : trouver la frontière qui sépare les classes avec la plus grande marge.
- **Intuition** : parmi toutes les frontières possibles, on retient celle la plus éloignée des points les plus proches de chaque classe (les vecteurs de support) ; un noyau permet des frontières courbes.
- **Étapes** : standardiser → résoudre le problème d'optimisation sous contraintes de marge → ne conserver que les vecteurs de support → classer avec le signe de la fonction de décision.
- **Formule** : $\displaystyle \min_{w,b,\xi}\tfrac12\|w\|^2+C\sum_i\xi_i$ sous $y_i(w^\top\phi(x_i)+b)\ge1-\xi_i$ ; $w$ = normale de l'hyperplan, $\|w\|^{-1}$ = marge, $\xi_i\ge0$ = violations tolérées, $C$ = poids de ces violations, $\phi$ = transformation implicite du noyau ; noyau RBF : $K(x,x')=e^{-\gamma\|x-x'\|^2}$.
- **Paramètres / hyperparamètres** : paramètres = $w,b$ (ou coefficients des vecteurs de support) ; hyperparamètres = `C` (1,0), `kernel` (`rbf`), `gamma`.
- **Avantage 1** : efficace en dimension élevée et avec peu d'observations.
- **Avantage 2** : frontières non linéaires flexibles grâce aux noyaux.
- **Limite 1** : coût élevé sur de grands échantillons (au moins quadratique) et sensibilité à l'échelle et à $C$, $\gamma$.
- **Limite 2** : pas de probabilités directes (calibration supplémentaire) et peu interprétable avec un noyau.

### Schéma explicatif

**Comment lire ce schéma.** Le trait plein est la frontière de décision ; les pointillés gris sont les bords de la marge. Seuls les points cerclés en vert (vecteurs de support), situés sur ou dans la marge, déterminent la frontière : déplacer les autres ne change rien. (Schéma en noyau linéaire pour la clarté ; l'exemple du notebook utilise le noyau RBF.)

```python
def schema():
    from sklearn.svm import SVC
    g = np.random.default_rng(11); X = np.r_[g.normal([-1.8, -1.5], .8, (30, 2)), g.normal([1.8, 1.6], .8, (30, 2))]; y = np.r_[np.zeros(30), np.ones(30)].astype(int)
    m = SVC(kernel='linear', C=1).fit(X, y)
    gx, gy = np.meshgrid(np.linspace(-4, 4, 200), np.linspace(-4, 4, 200)); z = m.decision_function(np.c_[gx.ravel(), gy.ravel()]).reshape(gx.shape)
    fig, ax = plt.subplots(figsize=(6, 5))
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolor='k', s=26)
    ax.contour(gx, gy, z, levels=[-1, 0, 1], colors=['gray', 'k', 'gray'], linestyles=['--', '-', '--'])
    ax.scatter(*m.support_vectors_.T, s=140, facecolors='none', edgecolors='#27ae60', lw=2, label='vecteurs de support')
    ax.plot([], [], 'k-', label='frontière $f(x)=0$'); ax.plot([], [], '--', c='gray', label='bords de marge $f(x)=\\pm1$')
    ax.set_xlabel('$x_1$'); ax.set_ylabel('$x_2$'); ax.set_title('SVM linéaire : maximiser la marge'); ax.legend(fontsize=8, loc='upper left'); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : classer des entreprises en solides/fragiles à partir de ratios financiers (échantillons de taille modeste).
- **Cas d'usage 2 (commerce)** : classification de textes (avis clients positifs/négatifs) en grande dimension.
- **Données et prérequis** : variables numériques standardisées, taille de l'échantillon raisonnable (quelques milliers), réglage de `C` et `gamma` par validation croisée.
- **Cas peu approprié** : très grands jeux de données (centaines de milliers de lignes) ou besoin de probabilités fiables sans calibration.
- **Métriques** : F1, précision, rappel, ROC-AUC (via la fonction de décision), nombre de vecteurs de support.

### Analyse de l'expérience Python

```python
from sklearn.svm import SVC
modele=make_pipeline(StandardScaler(),SVC(kernel='rbf',C=1.0,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

**Commentaire.**

**Entrées** : mêmes données ; pipeline `StandardScaler` + `SVC(kernel='rbf', C=1.0)`. **Sortie** : F1 = **0,898**.
**Lecture** : bon score, proche de KNN (0,889) et de l'arbre (0,891), supérieur à la régression logistique (0,811) : la frontière non linéaire du noyau RBF aide. Inférieur toutefois à la forêt (0,918) et au boosting (0,950).
**Limites** : `C` et `gamma` n'ont pas été optimisés ; l'écart avec les autres méthodes n'est pas significatif sur 100 observations.

## 14. Gradient Boosting — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Gradient Boosting pour la classification (*Gradient Boosting Classifier*, version `HistGradientBoostingClassifier`) — **famille** : classification supervisée (ensemble séquentiel) ; **objectif** : améliorer progressivement un classifieur en corrigeant ses erreurs.
- **Intuition** : chaque petit arbre est entraîné pour corriger la direction d'erreur (gradient de la perte) du modèle actuel.
- **Étapes** : initialiser par le log-odds de la classe → à chaque itération, calculer le gradient de la log-perte, ajuster un arbre dessus, l'ajouter avec un petit pas → transformer le score final en probabilité.
- **Formule** : $F_m(x)=F_{m-1}(x)+\nu\,h_m(x)$ et $P(y=1\mid x)=\sigma(F_M(x))$ ; $h_m$ = arbre ajusté sur $-\partial L/\partial F$ (gradient négatif de la log-perte $L$), $\nu$ = taux d'apprentissage, $\sigma$ = sigmoïde.
- **Paramètres / hyperparamètres** : paramètres = arbres $h_m$ ; hyperparamètres = `max_iter` (80), `learning_rate`, `max_depth`, `l2_regularization`, arrêt précoce.
- **Avantage 1** : souvent la meilleure précision sur données tabulaires.
- **Avantage 2** : la version à histogrammes est rapide et gère les valeurs manquantes.
- **Limite 1** : nombreux hyperparamètres interdépendants et risque de surapprentissage.
- **Limite 2** : modèle opaque, entraînement séquentiel.

### Schéma explicatif

**Comment lire ce schéma.** De gauche à droite, le même jeu de données (deux croissants) et la frontière du modèle après 1, 5 puis 60 arbres : la frontière, d'abord grossière (un seul découpage), s'affine et épouse la forme des classes à mesure que chaque arbre corrige les erreurs restantes.

```python
def schema():
    from sklearn.ensemble import GradientBoostingClassifier
    from sklearn.datasets import make_moons
    X, y = make_moons(120, noise=.28, random_state=0)
    fig, ax = plt.subplots(1, 3, figsize=(13, 3.8))
    for a, m in zip(ax, [1, 5, 60]):
        frontiere(a, GradientBoostingClassifier(n_estimators=m, max_depth=2, learning_rate=.3, random_state=0).fit(X, y), X, y, f'après {m} arbre(s)')
    plt.tight_layout(); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : prédiction du défaut de paiement sur des données de crédit à la consommation.
- **Cas d'usage 2 (marketing)** : ciblage de campagnes (probabilité d'achat) avec de nombreuses variables comportementales.
- **Données et prérequis** : tabulaires, assez d'observations, validation pour fixer le nombre d'itérations.
- **Cas peu approprié** : petits échantillons (surapprentissage) ou exigences d'interprétabilité forte.
- **Métriques** : F1, rappel, précision, ROC-AUC, PR-AUC, log-perte.

### Analyse de l'expérience Python

```python
from sklearn.ensemble import HistGradientBoostingClassifier
modele=HistGradientBoostingClassifier(max_iter=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : AdaBoost (*Adaptive Boosting*) — **famille** : classification supervisée (boosting) ; **objectif** : combiner de nombreux classifieurs faibles (souches) en un classifieur fort.
- **Intuition** : à chaque tour, on augmente le poids des observations mal classées pour forcer le tour suivant à s'y concentrer.
- **Étapes** : poids égaux → pour $t=1..T$ : entraîner une souche pondérée, mesurer son erreur, calculer son poids $\alpha_t$, réévaluer les poids des observations → vote pondéré.
- **Formule** : $\varepsilon_t=\sum_iw_i\mathbf 1[h_t(x_i)\ne y_i]$, $\alpha_t=\tfrac12\ln\frac{1-\varepsilon_t}{\varepsilon_t}$, $w_i\leftarrow w_i\,e^{-\alpha_ty_ih_t(x_i)}$ (puis normalisation), $H(x)=\mathrm{sign}\big(\sum_t\alpha_th_t(x)\big)$ ; $y_i\in\{-1,+1\}$, $h_t$ = classifieur faible du tour $t$.
- **Paramètres / hyperparamètres** : paramètres = souches $h_t$ et poids $\alpha_t$ ; hyperparamètres = `n_estimators` (70), `learning_rate`, classifieur faible de base.
- **Avantage 1** : simple, peu de réglages, bon gain par rapport à un classifieur faible.
- **Avantage 2** : s'applique avec tout classifieur faible.
- **Limite 1** : très sensible au bruit et aux valeurs aberrantes (leurs poids explosent).
- **Limite 2** : entraînement séquentiel et performance souvent dépassée par le gradient boosting.

### Schéma explicatif

**Comment lire ce schéma.** Chaque graphique est un tour de boosting. La taille d'un point est proportionnelle à son poids ; la frontière est la souche (un seul seuil) de ce tour ; les croix noires sont les points mal classés. On voit que les points mal classés au tour 1 grossissent au tour 2 : le nouvel apprenant est forcé de les traiter.

```python
def schema():
    from sklearn.tree import DecisionTreeClassifier
    g = np.random.default_rng(12); X = np.r_[g.normal([-1, -1], 1.1, (25, 2)), g.normal([1, 1], 1.1, (25, 2))]; y = np.r_[-np.ones(25), np.ones(25)].astype(int)
    w = np.ones(50)/50; fig, ax = plt.subplots(1, 3, figsize=(13, 3.9))
    for t in range(3):
        s = DecisionTreeClassifier(max_depth=1, random_state=0).fit(X, y, sample_weight=w); p = s.predict(X)
        err = w[p != y].sum(); a = .5*np.log((1 - err)/err)
        frontiere(ax[t], s, X, y, f'tour {t+1} : erreur pondérée={err:.2f}, $\\alpha$={a:.2f}')
        ax[t].collections[-1].set_sizes(3500*w); ax[t].scatter(X[p != y, 0], X[p != y, 1], marker='x', c='k', s=40)
        w = w*np.exp(-a*y*p); w /= w.sum()
    plt.tight_layout(); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : classification de transactions (suspectes / normales) avec des règles simples combinées.
- **Cas d'usage 2 (commerce)** : détection de clients à fort risque d'impayé sur données propres et peu bruitées.
- **Données et prérequis** : données propres (peu de bruit d'étiquettes), classifieurs faibles simples.
- **Cas peu approprié** : données avec beaucoup d'erreurs d'étiquetage ou de valeurs aberrantes.
- **Métriques** : F1, précision, rappel, ROC-AUC ; erreur par nombre de tours.

### Analyse de l'expérience Python

```python
from sklearn.ensemble import AdaBoostClassifier
modele=AdaBoostClassifier(n_estimators=70,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

**Commentaire.**

**Entrées** : mêmes données ; **hyperparamètres** : `n_estimators=70`, `random_state=42` (souche de profondeur 1 par défaut). **Sortie** : F1 = **0,827**.
**Lecture** : inférieur aux forêts, au boosting par gradient (0,950) et au SVM, mais supérieur à la régression logistique (0,811) et à Naive Bayes (0,755). Des souches (un seul seuil) ont une capacité limitée même combinées.
**Limites** : aucun réglage du nombre de tours ni du pas ; on ne peut pas attribuer l'écart avec le boosting par gradient à l'algorithme plutôt qu'au réglage.

## 16. XGBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : XGBoost (*eXtreme Gradient Boosting*) — **famille** : classification (et régression) supervisée, boosting d'arbres régularisé ; **objectif** : une implémentation de boosting par gradient rapide et fortement régularisée.
- **Intuition** : comme le gradient boosting, on ajoute des arbres, mais chaque arbre est construit avec une approximation du second ordre de la perte et une pénalité explicite sur sa complexité.
- **Étapes** : calculer gradients $g_i$ et hessiens $h_i$ → construire l'arbre qui maximise le gain → ajouter l'arbre avec le pas $\eta$ → répéter.
- **Formule** : objectif $\displaystyle \mathcal L=\sum_i\ell(y_i,\hat y_i)+\sum_k\big[\gamma T_k+\tfrac12\lambda\|w_k\|^2\big]$ ; gain d'une coupure $\tfrac12\Big[\frac{G_L^2}{H_L+\lambda}+\frac{G_R^2}{H_R+\lambda}-\frac{(G_L+G_R)^2}{H_L+H_R+\lambda}\Big]-\gamma$ ; $G,H$ = sommes des $g_i,h_i$ à gauche/droite, $T_k$ = nombre de feuilles, $w_k$ = valeurs des feuilles, $\gamma,\lambda$ = pénalités.
- **Paramètres / hyperparamètres** : paramètres = arbres ; hyperparamètres = `n_estimators` (80), `max_depth` (3), `learning_rate` ($\eta$), `subsample`, `colsample_bytree`, $\lambda$, $\gamma$.
- **Avantage 1** : très performant en compétitions tabulaires, régularisation intégrée, gère les valeurs manquantes.
- **Avantage 2** : rapide (parallélisation, histogrammes) et passe à l'échelle.
- **Limite 1** : beaucoup d'hyperparamètres à régler et risque de surapprentissage sans arrêt précoce.
- **Limite 2** : modèle opaque et dépendance à une bibliothèque externe (installation).

### Schéma explicatif

**Comment lire ce schéma.** Lire de gauche à droite : à partir de la prédiction courante, on mesure l'erreur par le gradient et le hessien de la perte (pentes et courbures), on construit un arbre qui réduit au mieux cette erreur tout en payant une pénalité de complexité, puis on l'ajoute au modèle avec un petit pas $\eta$. La flèche du bas indique la boucle de répétition.

```python
def schema():
    fig, ax = plt.subplots(figsize=(12, 3.8)); toile(ax, (0, 12.6), (0, 4), 'XGBoost : une itération de boosting (répétée $M$ fois)')
    boite(ax, 1.5, 2, 'Données\n$(x_i,y_i)$\nprédiction courante\n$\\hat y_i^{(t-1)}$', 2.6, 1.6, '#eaf2f8')
    boite(ax, 4.6, 2, 'Gradients $g_i$\net hessiens $h_i$\nde la perte', 2.6, 1.6, '#fdebd0')
    boite(ax, 7.7, 2, 'Nouvel arbre $f_t$\nmaximisant le gain\n(pénalités $\\lambda,\\gamma$)', 2.8, 1.6, '#d5f5e3')
    boite(ax, 10.7, 2, 'Mise à jour\n$\\hat y^{(t)}=\\hat y^{(t-1)}+\\eta f_t$', 2.6, 1.6, '#fcf3cf')
    for a, b in [(2.8, 3.3), (5.9, 6.3), (9.1, 9.4)]: fleche(ax, a, 2, b, 2)
    fleche(ax, 10.7, 1.1, 1.5, 1.1, 'on recommence avec la nouvelle prédiction', couleur='#7f8c8d', decalage=-.35, courbe=-.15)
    ax.text(6, 3.6, 'prédiction finale = somme de tous les arbres pondérés par $\\eta$', ha='center', style='italic', fontsize=9); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : modèle de risque de crédit ou de détection de fraude sur de grandes bases de transactions.
- **Cas d'usage 2 (logistique)** : prévoir les retards de livraison à partir de nombreuses variables hétérogènes.
- **Données et prérequis** : tabulaires, bibliothèque `xgboost` installée, validation pour l'arrêt précoce.
- **Cas peu approprié** : très petits jeux de données ou exigence d'explication simple aux clients.
- **Métriques** : F1, rappel, précision, ROC-AUC, PR-AUC, log-perte.

### Analyse de l'expérience Python

```python
from xgboost import XGBClassifier
modele=XGBClassifier(n_estimators=80,max_depth=3,random_state=GRAINE,eval_metric='logloss').fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

**Commentaire.**

**Entrées** : mêmes données ; **hyperparamètres** : `n_estimators=80`, `max_depth=3`, `eval_metric='logloss'`, `random_state=42`. **Sortie** : la cellule affiche le F1 sur le test.
**Résultat** : cette cellule dépend de la bibliothèque `xgboost`, installée par la première cellule du notebook ; la valeur exacte est à relever dans votre exécution (je ne la reporte pas, car elle n'a pas pu être exécutée dans l'environnement où ce texte a été rédigé). On peut s'attendre à un ordre de grandeur proche du `HistGradientBoostingClassifier` (0,950), les deux étant des boostings d'arbres, mais ce n'est qu'une attente à vérifier.
**Ce que la sortie permet / ne permet pas** : comparer à 100 observations de test, avec un seul découpage, ne permet pas de conclure que XGBoost est supérieur ; recopiez la valeur obtenue et comparez-la prudemment.

## 17. K-means

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : K-means (*K-means clustering*) — **famille** : clustering non supervisé ; **objectif** : regrouper les observations en $K$ groupes compacts.
- **Intuition** : chaque groupe est représenté par son centre (centroïde) ; chaque point appartient au centre le plus proche.
- **Étapes** : choisir $K$ centres initiaux → assigner chaque point au centre le plus proche → recalculer chaque centre comme la moyenne de son groupe → répéter jusqu'à stabilité.
- **Critère** : $\displaystyle \min_{C,\mu}\sum_{k=1}^{K}\sum_{x_i\in C_k}\|x_i-\mu_k\|^2$ (inertie) ; $C_k$ = groupe $k$, $\mu_k$ = centroïde du groupe $k$.
- **Paramètres / hyperparamètres** : paramètres = centroïdes $\mu_k$ et affectations ; hyperparamètres = $K$ (3), `n_init` (10 relances), initialisation (`k-means++`).
- **Avantage 1** : simple, rapide et passe à l'échelle.
- **Avantage 2** : résultats faciles à interpréter (profil moyen de chaque groupe).
- **Limite 1** : il faut fixer $K$ à l'avance et il suppose des groupes sphériques de tailles comparables.
- **Limite 2** : sensible à l'initialisation, aux échelles des variables et aux valeurs aberrantes.

### Schéma explicatif

**Comment lire ce schéma.** Les croix rouges sont les centroïdes, les couleurs les groupes. À gauche, trois centres de départ (les trois premiers points) et l'affectation des points au centre le plus proche ; au milieu, les centres ont été recalculés comme moyennes des groupes ; à droite, après plusieurs répétitions, les centres ne bougent plus : l'algorithme a convergé.

```python
def schema():
    from sklearn.datasets import make_blobs
    X, _ = make_blobs(90, centers=3, cluster_std=1.1, random_state=3)
    mu = X[[0, 1, 2]].copy(); fig, ax = plt.subplots(1, 3, figsize=(13, 3.9)); etapes = {0: 'initialisation des centres', 1: 'après 1 itération', 2: 'convergence'}
    for it in range(3):
        d = ((X[:, None] - mu[None])**2).sum(2); lab = d.argmin(1)
        if it == 2:
            for _ in range(30):
                mu = np.array([X[lab == k].mean(0) for k in range(3)]); lab = ((X[:, None] - mu[None])**2).sum(2).argmin(1)
        ax[it].scatter(X[:, 0], X[:, 1], c=lab, cmap='viridis', s=20, edgecolor='k', lw=.3); ax[it].scatter(*mu.T, marker='X', s=200, c='red', edgecolor='k', label='centroïdes')
        ax[it].set_title(etapes[it], fontsize=9); ax[it].legend(fontsize=8); ax[it].set_xlabel('$x_1$'); ax[it].set_ylabel('$x_2$')
        if it == 0: mu = np.array([X[lab == k].mean(0) for k in range(3)])
        if it == 1: pass
    plt.tight_layout(); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (marketing)** : segmentation de clientèle (méthode RFM : récence, fréquence, montant).
- **Cas d'usage 2 (finance / logistique)** : regrouper des actions par profil rendement-risque, ou des entrepôts par volume et distance.
- **Données et prérequis** : variables numériques standardisées, groupes approximativement sphériques, choix de $K$ (méthode du coude, silhouette).
- **Cas peu approprié** : groupes de forme allongée/imbriquée ou de densités très différentes (DBSCAN ou GMM plus adaptés).
- **Métriques** : coefficient de silhouette, inertie, indice de Davies-Bouldin, stabilité entre relances ; ARI si des étiquettes de référence existent.

### Analyse de l'expérience Python

```python
from sklearn.cluster import KMeans
modele=KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(X_clu)
print('Silhouette:',silhouette_score(X_clu,modele.labels_))
plt.scatter(X_clu[:,0],X_clu[:,1],c=modele.labels_,s=12);plt.title('K-means');plt.show()
```

**Commentaire.**

**Entrées** : `X_clu` (240 points en 2D, 3 groupes générés par `make_blobs`) ; **hyperparamètres** : `n_clusters=3`, `n_init=10`, `random_state=42`. **Sortie** : silhouette = **0,869** et un nuage de points coloré par groupe.
**Lecture** : une silhouette proche de 1 indique des groupes compacts et bien séparés ; 0,87 est élevé, ce qui est normal car les trois blobs sont bien distincts (écart-type 0,85 pour des centres éloignés).
**Limites** : on a donné $K=3$ en sachant qu'il y a 3 groupes ; la silhouette ne prouve pas que ce $K$ soit le bon (il faudrait la comparer pour plusieurs $K$). Les données étant synthétiques, ce score est un maximum optimiste par rapport aux données réelles.

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Classification ascendante hiérarchique (*Agglomerative hierarchical clustering*, CAH) — **famille** : clustering non supervisé ; **objectif** : construire une hiérarchie de regroupements imbriqués.
- **Intuition** : on part de chaque point isolé et on fusionne successivement les deux groupes les plus proches ; l'arbre de fusions (dendrogramme) montre la structure à toutes les échelles.
- **Étapes** : calculer les distances → fusionner les deux groupes les plus proches selon un critère de lien → mettre à jour les distances → couper l'arbre au niveau souhaité.
- **Critère** (lien de Ward, par défaut) : on fusionne la paire $(A,B)$ qui minimise l'augmentation d'inertie $\displaystyle \Delta=\frac{|A||B|}{|A|+|B|}\|\mu_A-\mu_B\|^2$ ; $|A|$ = effectif du groupe $A$, $\mu_A$ = son centre.
- **Paramètres / hyperparamètres** : pas de paramètres appris ; hyperparamètres = `n_clusters` (3) ou seuil de distance, `linkage` (`ward`, `average`, `complete`, `single`), métrique.
- **Avantage 1** : pas besoin de fixer $K$ à l'avance (le dendrogramme guide le choix).
- **Avantage 2** : détecte des structures à plusieurs échelles ; déterministe.
- **Limite 1** : coût en $O(n^2)$ en mémoire et plus : inadapté aux très grands jeux de données.
- **Limite 2** : une fusion est irréversible et sensible au choix du lien et aux valeurs aberrantes.

### Schéma explicatif

**Comment lire ce schéma.** Chaque feuille en bas est une observation. Les traits en U montrent les fusions successives ; la hauteur d'un U est la distance à laquelle deux groupes sont fusionnés. La ligne pointillée « coupe » l'arbre : le nombre de branches qu'elle croise est le nombre de groupes retenus (3 ici, en couleurs).

```python
def schema():
    from scipy.cluster.hierarchy import linkage, dendrogram
    from sklearn.datasets import make_blobs
    X, _ = make_blobs(24, centers=3, cluster_std=.9, random_state=2); Z = linkage(X, 'ward'); seuil = Z[-3, 2] + .01
    fig, ax = plt.subplots(figsize=(8, 4)); dendrogram(Z, color_threshold=seuil, ax=ax, no_labels=True)
    ax.axhline(seuil, c='k', ls='--', label='coupure → 3 groupes'); ax.set_xlabel('observations (feuilles)'); ax.set_ylabel('distance de fusion (Ward)')
    ax.set_title('Dendrogramme : histoire des fusions'); ax.legend(); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : regrouper des pays ou des actifs selon leurs indicateurs/corrélations et visualiser la hiérarchie des ressemblances.
- **Cas d'usage 2 (commerce)** : construire une taxonomie de produits ou de clients à partir de leurs profils d'achat.
- **Données et prérequis** : variables numériques standardisées, quelques centaines à milliers d'observations.
- **Cas peu approprié** : grandes bases (≫ 10 000 lignes) où la matrice des distances devient trop volumineuse.
- **Métriques** : silhouette, corrélation cophénétique, stabilité selon le lien ; ARI si étiquettes de référence.

### Analyse de l'expérience Python

```python
from sklearn.cluster import AgglomerativeClustering
etiquettes=AgglomerativeClustering(n_clusters=3).fit_predict(X_clu)
print('Silhouette:',silhouette_score(X_clu,etiquettes))
```

**Commentaire.**

**Entrées** : `X_clu` ; **hyperparamètres** : `n_clusters=3`, lien de Ward (défaut). **Sortie** : silhouette = **0,869**.
**Lecture** : valeur identique, à toutes les décimales affichées, à celle de K-means : les deux méthodes ont très probablement produit les mêmes groupes, ce qui est normal sur trois blobs bien séparés.
**Limites** : une silhouette identique ne suffit pas à *prouver* l'égalité des partitions (il faudrait comparer les étiquettes, par exemple avec l'ARI) ; et sur des données moins bien séparées, K-means et la CAH donneraient des résultats différents.

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : DBSCAN (*Density-Based Spatial Clustering of Applications with Noise*) — **famille** : clustering non supervisé par densité (et détection de bruit) ; **objectif** : trouver des groupes de forme quelconque et isoler les points isolés.
- **Intuition** : un groupe est une région dense de points ; les points dans des zones vides sont du bruit.
- **Étapes** : pour chaque point, compter les voisins dans un rayon $\varepsilon$ → points « cœurs » s'ils ont au moins `min_samples` voisins → relier les cœurs voisins pour former les groupes → rattacher les points « frontières » ; le reste est du bruit (étiquette −1).
- **Critère** : $N_\varepsilon(x)=\{x'\,:\,\|x-x'\|\le\varepsilon\}$ ; $x$ est un point cœur si $|N_\varepsilon(x)|\ge \text{min\_samples}$ ; $\varepsilon$ = rayon de voisinage.
- **Paramètres / hyperparamètres** : pas de paramètres appris ; hyperparamètres = `eps` ($\varepsilon=0{,}8$) et `min_samples` (5).
- **Avantage 1** : pas besoin de fixer le nombre de groupes ; formes arbitraires.
- **Avantage 2** : identifie naturellement le bruit/les anomalies.
- **Limite 1** : très sensible à `eps` et mal adapté aux groupes de densités différentes.
- **Limite 2** : difficile en grande dimension (les distances perdent leur sens) et nécessite des variables standardisées.

### Schéma explicatif

**Comment lire ce schéma.** Deux demi-lunes (formes non sphériques) sont retrouvées là où K-means échouerait. Les gros points sont des cœurs (au moins `min_samples` voisins dans le rayon $\varepsilon$ — cercle rouge), les petits des points frontières, et les croix noires du bruit isolé.

```python
def schema():
    from sklearn.cluster import DBSCAN
    from sklearn.datasets import make_moons
    g = np.random.default_rng(13); X = np.r_[make_moons(200, noise=.06, random_state=0)[0], g.uniform(-1.3, 2.3, (12, 2))*[1, .9] + [0, -.2]]
    m = DBSCAN(eps=.15, min_samples=5).fit(X); lab = m.labels_; coeur = np.zeros(len(X), bool); coeur[m.core_sample_indices_] = True
    fig, ax = plt.subplots(figsize=(7, 4.4))
    for k, c in [(0, '#2980b9'), (1, '#27ae60')]:
        ax.scatter(*X[(lab == k) & coeur].T, c=c, s=34, edgecolor='k', lw=.4, label=f'groupe {k} : points cœurs')
        ax.scatter(*X[(lab == k) & ~coeur].T, c=c, s=12, label=f'groupe {k} : points frontières')
    ax.scatter(*X[lab == -1].T, marker='x', c='k', s=40, label='bruit (étiquette −1)')
    i = m.core_sample_indices_[0]; ax.add_patch(Circle(X[i], .15, fill=False, ec='red', lw=1.5)); ax.annotate(r'rayon $\varepsilon$', X[i] + [.1, .12], color='red')
    ax.set_aspect('equal'); ax.set_title('DBSCAN : groupes denses de forme libre et bruit'); ax.set_xlabel('$x_1$'); ax.set_ylabel('$x_2$'); ax.legend(fontsize=7, loc='upper right'); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : repérer des transactions isolées (potentielles fraudes) à côté de groupes de comportements normaux.
- **Cas d'usage 2 (logistique)** : regrouper des points de livraison géolocalisés en zones denses de forme irrégulière.
- **Données et prérequis** : variables numériques standardisées ; choix de `eps` (graphe des distances au $k$-ième voisin).
- **Cas peu approprié** : densités très hétérogènes ou haute dimension.
- **Métriques** : nombre de groupes et de points de bruit, silhouette (hors bruit), stabilité selon `eps` ; ARI si étiquettes de référence.

## 20. Mélanges gaussiens

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Mélanges gaussiens (*Gaussian Mixture Models*, GMM) — **famille** : clustering non supervisé probabiliste ; **objectif** : modéliser les données comme un mélange de lois normales et affecter chaque point à ses composantes avec une probabilité.
- **Intuition** : chaque groupe est une « cloche » gaussienne avec son centre, sa forme et son poids ; un point peut appartenir « un peu » à plusieurs groupes.
- **Étapes** : initialiser les composantes → étape E : calculer les probabilités d'appartenance (responsabilités) → étape M : réestimer poids, moyennes, covariances → répéter jusqu'à convergence (algorithme EM).
- **Formule** : $\displaystyle p(x)=\sum_{k=1}^{K}\pi_k\,\mathcal N(x;\mu_k,\Sigma_k)$ ; $\pi_k$ = poids de la composante $k$ ($\sum\pi_k=1$), $\mu_k$ = moyenne, $\Sigma_k$ = matrice de covariance ; responsabilité $r_{ik}=\pi_k\mathcal N(x_i;\mu_k,\Sigma_k)/p(x_i)$.
- **Paramètres / hyperparamètres** : paramètres = $\pi_k,\mu_k,\Sigma_k$ ; hyperparamètres = nombre de composantes (3), `covariance_type`, initialisation ; le critère BIC/AIC aide à choisir $K$.
- **Avantage 1** : groupes de formes elliptiques différentes et affectation souple (probabilités).
- **Avantage 2** : cadre probabiliste (sélection de $K$ par BIC, détection d'anomalies par faible densité).
- **Limite 1** : suppose des composantes gaussiennes et peut converger vers un optimum local.
- **Limite 2** : nombreux paramètres en grande dimension (covariances) donc risque d'instabilité.

### Schéma explicatif

**Comment lire ce schéma.** Chaque ellipse est une composante gaussienne (trait plein = 1 écart-type, pointillé = 2) : sa position est la moyenne, sa forme et son inclinaison viennent de la covariance. La couleur d'un point mélange les couleurs des composantes selon ses probabilités d'appartenance : les points proches d'une frontière sont de teinte intermédiaire.

```python
def schema():
    from sklearn.mixture import GaussianMixture
    g = np.random.default_rng(14); A = np.array([[1.6, .8], [.0, .5]])
    X = np.r_[g.normal(size=(100, 2)) @ A + [0, 0], g.normal(size=(100, 2)) @ A.T + [5, 3], g.normal(size=(100, 2))*.6 + [0, 5]]
    m = GaussianMixture(3, random_state=0).fit(X); p = m.predict_proba(X); fig, ax = plt.subplots(figsize=(6.6, 5))
    ax.scatter(X[:, 0], X[:, 1], c=p, s=18, edgecolor='k', lw=.2, label='couleur = mélange des probabilités d\'appartenance')
    for mu, S in zip(m.means_, m.covariances_):
        v, w = np.linalg.eigh(S); ang = np.degrees(np.arctan2(w[1, 1], w[0, 1]))
        for nsig, ls in [(1, '-'), (2, '--')]: ax.add_patch(Ellipse(mu, 2*nsig*np.sqrt(v[1]), 2*nsig*np.sqrt(v[0]), angle=ang, fill=False, ec='k', ls=ls))
    ax.scatter(*m.means_.T, marker='X', s=160, c='red', edgecolor='k', label='moyennes $\\mu_k$'); ax.set_xlabel('$x_1$'); ax.set_ylabel('$x_2$')
    ax.set_title('Mélange gaussien : ellipses à 1σ (trait plein) et 2σ (pointillés)'); ax.legend(fontsize=7, loc='lower right'); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : modéliser des rendements par régimes de marché (calme/turbulent) avec affectation probabiliste.
- **Cas d'usage 2 (marketing)** : segmentation « douce » de clients pouvant relever de plusieurs profils.
- **Données et prérequis** : variables continues, groupes approximativement elliptiques ; assez d'observations pour estimer les covariances.
- **Cas peu approprié** : variables catégorielles ou très asymétriques, petit échantillon en grande dimension.
- **Métriques** : log-vraisemblance, BIC/AIC, silhouette, stabilité ; ARI si étiquettes de référence.

### Analyse de l'expérience Python

```python
from sklearn.mixture import GaussianMixture
modele=GaussianMixture(n_components=3,random_state=GRAINE).fit(X_clu)
print('Poids:',modele.weights_,'Silhouette:',silhouette_score(X_clu,modele.predict(X_clu)))
```

**Commentaire.**

**Entrées** : `X_clu` ; **hyperparamètres** : `n_components=3`, `random_state=42`. **Sorties** : poids = **[0,333 ; 0,333 ; 0,333]** et silhouette = **0,869**.
**Lecture** : les poids égaux à 1/3 correspondent à trois groupes de 80 points (240/3), ce qui est cohérent avec la génération des données. La silhouette est identique à celle de K-means et de la CAH : les trois méthodes retrouvent la même structure bien séparée.
**Limites** : la silhouette ne mesure pas la qualité probabiliste (la vraisemblance, le BIC ne sont pas affichés) et une partition identique n'est pas démontrée par un simple score.

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Analyse en composantes principales (*Principal Component Analysis*, ACP/PCA) — **famille** : réduction de dimension non supervisée ; **objectif** : résumer beaucoup de variables corrélées par quelques axes qui conservent le maximum de variance.
- **Intuition** : on cherche les directions dans lesquelles les données sont le plus étalées ; projeter sur les premières donne la meilleure vue en peu de dimensions.
- **Étapes** : centrer (et réduire) les variables → calculer la matrice de covariance → extraire ses vecteurs propres → projeter sur les $q$ premiers.
- **Formule** : $\Sigma v_j=\lambda_jv_j$ ; coordonnées $Z=XV_q$ ; part de variance expliquée par l'axe $j$ : $\lambda_j/\sum_{l=1}^p\lambda_l$ ; $\Sigma$ = covariance, $v_j$ = axe principal $j$, $\lambda_j$ = variance sur cet axe, $V_q$ = matrice des $q$ premiers axes.
- **Paramètres / hyperparamètres** : paramètres = axes $v_j$ et valeurs propres $\lambda_j$ ; hyperparamètre = nombre de composantes (`n_components`, ici 2), standardisation préalable.
- **Avantage 1** : méthode déterministe, rapide, qui supprime la redondance et le bruit.
- **Avantage 2** : permet la visualisation et la décorrélation des variables avant d'autres modèles.
- **Limite 1** : uniquement linéaire (ne déplie pas des structures courbes).
- **Limite 2** : sensible aux échelles et aux valeurs aberrantes ; les axes sont difficiles à interpréter.

### Schéma explicatif

**Comment lire ce schéma.** Le nuage gris est allongé selon une diagonale. L'axe 1 (rouge) suit la direction d'étalement maximal ; l'axe 2 (vert) lui est orthogonal. Les segments gris relient quelques points à leur projection sur l'axe 1 (points bleus) : l'ACP choisit l'axe qui rend ces segments les plus courts (perte d'information minimale).

```python
def schema():
    g = np.random.default_rng(15); X = g.normal(size=(120, 2)) @ np.array([[2.2, 1.1], [0, .6]]); X = X - X.mean(0)
    v, w = np.linalg.eigh(np.cov(X.T)); w = w[:, ::-1]; v = v[::-1]; proj = (X @ w[:, [0]]) @ w[:, [0]].T
    fig, ax = plt.subplots(figsize=(6.2, 5)); ax.scatter(X[:, 0], X[:, 1], s=16, c='gray', label='observations')
    ax.plot(*np.c_[X, proj].reshape(-1, 2, 2).transpose(2, 0, 1)[:, :30], c='#aab7b8', lw=.6); ax.scatter(*proj.T, s=10, c='#2980b9', label='projections sur l\'axe 1')
    for j, c in enumerate(['#c0392b', '#27ae60']): ax.annotate('', 2.2*np.sqrt(v[j])*w[:, j], (0, 0), arrowprops=dict(arrowstyle='-|>', color=c, lw=2.5)); ax.text(*(2.3*np.sqrt(v[j])*w[:, j]), f'axe {j+1} ({v[j]/v.sum():.0%} de la variance)', color=c, fontsize=8)
    ax.set_aspect('equal'); ax.set_xlabel('variable 1'); ax.set_ylabel('variable 2'); ax.set_title('ACP : axes de variance maximale'); ax.legend(fontsize=8, loc='lower right'); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : extraire des facteurs communs d'une courbe de taux (niveau, pente, courbure) ou d'un panier d'actions.
- **Cas d'usage 2 (marketing)** : réduire des dizaines de réponses d'enquête à quelques dimensions de perception.
- **Données et prérequis** : variables numériques corrélées, standardisées si échelles différentes.
- **Cas peu approprié** : structure non linéaire (ex. spirale), variables catégorielles ou besoin d'interpréter chaque axe.
- **Métriques** : variance expliquée (cumulée), erreur de reconstruction, diagramme des éboulis.

### Analyse de l'expérience Python

```python
from sklearn.decomposition import PCA
X_vin=StandardScaler().fit_transform(load_wine().data)
modele=PCA(n_components=2).fit(X_vin)
projection=modele.transform(X_vin)
print('Variance expliquée:',modele.explained_variance_ratio_)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.show()
```

**Commentaire.**

**Entrées** : base `load_wine` (178 vins, 13 variables chimiques, réelle) standardisée ; **hyperparamètre** : `n_components=2`. **Sortie** : variance expliquée = **[0,362 ; 0,192]** et un nuage de points 2D.
**Lecture** : le premier axe porte 36,2 % de la variance et le second 19,2 %, soit **55,4 %** à eux deux : le plan 2D conserve un peu plus de la moitié de l'information des 13 variables.
**Limites** : 44,6 % de la variance est perdue ; le nuage n'est pas coloré par cultivar, donc il ne dit pas si les classes sont séparées. Il faudrait aussi regarder les contributions des variables pour interpréter les axes.

## 22. t-SNE

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : t-SNE (*t-distributed Stochastic Neighbor Embedding*) — **famille** : réduction de dimension non linéaire, surtout pour la visualisation ; **objectif** : représenter en 2D ou 3D des données de grande dimension en conservant les voisinages locaux.
- **Intuition** : on traduit les distances entre points en probabilités de « voisinage », puis on place les points en 2D pour que ces probabilités soient reproduites au mieux.
- **Étapes** : calculer les affinités $p_{ij}$ en haute dimension → initialiser les points en 2D → calculer les affinités $q_{ij}$ (loi de Student) → minimiser la divergence KL par descente de gradient.
- **Formule** : $\displaystyle KL(P\|Q)=\sum_{i\ne j}p_{ij}\ln\frac{p_{ij}}{q_{ij}}$, avec $q_{ij}\propto(1+\|y_i-y_j\|^2)^{-1}$ ; $p_{ij}$ = similarité de $x_i$ et $x_j$ (gaussienne calibrée par la perplexité), $y_i$ = position 2D de l'observation $i$.
- **Paramètres / hyperparamètres** : paramètres = coordonnées 2D $y_i$ ; hyperparamètres = `perplexity` (20 ici), taux d'apprentissage, itérations, initialisation.
- **Avantage 1** : révèle des groupes locaux invisibles en ACP.
- **Avantage 2** : très répandu pour explorer des données complexes.
- **Limite 1** : les distances entre groupes et les tailles de groupes ne sont pas interprétables ; dépend de la perplexité et de la graine.
- **Limite 2** : coûteux sur de grands $n$ et sans fonction pour projeter de nouveaux points.

### Schéma explicatif

**Comment lire ce schéma.** Les mêmes 600 images de chiffres manuscrits (64 variables chacune, base réelle `load_digits`) sont projetées en 2D, colorées par chiffre. L'ACP (gauche) mélange beaucoup les classes ; t-SNE (droite) les sépare en îlots car il préserve les voisins proches. Attention : la distance entre îlots n'a pas de signification.

```python
def schema():
    from sklearn.datasets import load_digits
    from sklearn.decomposition import PCA
    from sklearn.manifold import TSNE
    d = load_digits(); idx = np.random.default_rng(0).choice(len(d.data), 600, replace=False); X, y = d.data[idx], d.target[idx]
    P = PCA(2, random_state=0).fit_transform(X); T = TSNE(perplexity=30, random_state=0, init='pca').fit_transform(X)
    fig, ax = plt.subplots(1, 2, figsize=(12, 4.8))
    for a, Z, t in zip(ax, [P, T], ['ACP (projection linéaire)', 't-SNE (conserve les voisinages)']):
        s = a.scatter(*Z.T, c=y, cmap='tab10', s=10); a.set_title(t); a.set_xlabel('dimension 1'); a.set_ylabel('dimension 2')
    fig.colorbar(s, ax=ax, label='chiffre manuscrit (0 à 9)', shrink=.85); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : visualiser des profils de clients ou de transactions en grande dimension pour repérer des groupes ou des cas atypiques.
- **Cas d'usage 2 (marketing)** : explorer des représentations vectorielles de produits ou de textes d'avis.
- **Données et prérequis** : variables numériques standardisées ; souvent précédé d'une ACP (30 à 50 composantes) ; $n$ jusqu'à quelques dizaines de milliers.
- **Cas peu approprié** : utilisation en entrée d'un modèle prédictif ou pour mesurer des distances entre groupes.
- **Métriques** : *trustworthiness* (conservation des voisinages), erreur KL finale, cohérence entre plusieurs graines et perplexités.

### Analyse de l'expérience Python

```python
from sklearn.manifold import TSNE
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=TSNE(n_components=2,perplexity=20,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('t-SNE');plt.show()
```

**Commentaire.**

**Entrées** : base `load_wine` standardisée ; **hyperparamètres** : `perplexity=20`, `random_state=42`. **Sortie** : un nuage de points 2D (aucune valeur numérique n'est affichée).
**Lecture** : on peut y repérer visuellement des regroupements de vins semblables. Aucune métrique n'est calculée : on ne peut donc ni quantifier la qualité de la projection ni la comparer à l'ACP.
**Limites** : le résultat dépend de la perplexité et de la graine ; les distances entre îlots ne sont pas interprétables ; sans coloration par cultivar, on ne peut pas dire si les groupes visibles correspondent aux vraies classes. À compléter par un score de voisinage (*trustworthiness*).

## 23. UMAP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : UMAP (*Uniform Manifold Approximation and Projection*) — **famille** : réduction de dimension non linéaire (visualisation, prétraitement) ; **objectif** : plonger les données dans un espace de faible dimension en conservant structure locale et, en partie, globale.
- **Intuition** : on suppose que les données vivent sur une surface courbe (variété) ; on construit un graphe de voisinage « flou » puis on cherche une disposition 2D dont le graphe est le plus semblable.
- **Étapes** : chercher les $k$ plus proches voisins → construire le graphe pondéré flou → initialiser la disposition (spectrale) → optimiser l'entropie croisée entre les graphes haute et basse dimension.
- **Formule** : minimisation de $\displaystyle \sum_{i\ne j}\Big[w_{ij}\ln\frac{w_{ij}}{v_{ij}}+(1-w_{ij})\ln\frac{1-w_{ij}}{1-v_{ij}}\Big]$ ; $w_{ij}$ = poids d'arête (appartenance floue) en haute dimension, $v_{ij}$ = poids correspondant dans l'espace 2D.
- **Paramètres / hyperparamètres** : paramètres = coordonnées 2D ; hyperparamètres = `n_neighbors` (15 ici), `min_dist`, `metric`, `n_components`.
- **Avantage 1** : plus rapide que t-SNE et transformation de nouveaux points possible (`transform`).
- **Avantage 2** : préserve généralement mieux la structure globale que t-SNE.
- **Limite 1** : sensible aux hyperparamètres et à la graine ; les distances restent d'interprétation prudente.
- **Limite 2** : bibliothèque externe (`umap-learn`) et fondements mathématiques plus techniques.

### Schéma explicatif

**Comment lire ce schéma.** À gauche : l'étape initiale d'UMAP, un graphe où chaque point est relié à ses plus proches voisins (traits bleus) ; c'est cette structure locale qu'on cherche à conserver. À droite : les trois étapes suivantes lues de gauche à droite (pondération des arêtes, disposition de départ, optimisation).

```python
def schema():
    from sklearn.neighbors import NearestNeighbors
    from sklearn.datasets import make_blobs
    X, _ = make_blobs(70, centers=3, cluster_std=1.0, random_state=5); idx = NearestNeighbors(n_neighbors=4).fit(X).kneighbors(X, return_distance=False)
    fig, ax = plt.subplots(1, 2, figsize=(12.5, 4.6), gridspec_kw={'width_ratios': [1, 1.2]})
    for i in range(len(X)):
        for j in idx[i, 1:]: ax[0].plot(*X[[i, j]].T, c='#85c1e9', lw=.7)
    ax[0].scatter(*X.T, s=18, c='#1f618d', zorder=3); ax[0].set_title('Étape 1 : graphe des 3 plus proches voisins'); ax[0].set_xlabel('$x_1$'); ax[0].set_ylabel('$x_2$')
    toile(ax[1], (0, 10), (0, 5), 'Étapes suivantes')
    boite(ax[1], 1.6, 2.5, 'Graphe flou\npondéré\n(poids $w_{ij}$)', 2.6, 1.8, '#eaf2f8'); boite(ax[1], 5, 2.5, 'Disposition\ninitiale 2D\n(spectrale)', 2.6, 1.8, '#fdebd0'); boite(ax[1], 8.4, 2.5, 'Optimisation\nde l\'entropie\ncroisée (min_dist)', 2.6, 1.8, '#d5f5e3')
    fleche(ax[1], 2.95, 2.5, 3.65, 2.5); fleche(ax[1], 6.35, 2.5, 7.05, 2.5); plt.tight_layout(); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : cartographier de grandes bases de clients ou d'instruments financiers et projeter de nouveaux clients sur la carte.
- **Cas d'usage 2 (marketing / commerce)** : visualiser des vecteurs de produits ou de documents sur des dizaines de milliers de points.
- **Données et prérequis** : variables numériques standardisées, métrique adaptée ; `umap-learn` installé.
- **Cas peu approprié** : échantillons très petits (le graphe de voisinage est peu fiable) ou interprétation quantitative des distances.
- **Métriques** : *trustworthiness*, conservation des $k$ voisins, stabilité entre graines et valeurs de `n_neighbors`.

### Analyse de l'expérience Python

```python
import umap
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=umap.UMAP(n_components=2,n_neighbors=15,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('UMAP');plt.show()
```

**Commentaire.**

**Entrées** : base `load_wine` standardisée ; **hyperparamètres** : `n_neighbors=15`, `n_components=2`, `random_state=42`. **Sortie** : un nuage de points 2D.

## 24. Isolation Forest

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Forêt d'isolement (*Isolation Forest*) — **famille** : détection d'anomalies non supervisée ; **objectif** : repérer les observations rares et atypiques.
- **Intuition** : une anomalie est « facile à isoler » : quelques coupures aléatoires suffisent pour la séparer des autres, alors qu'un point normal, noyé dans la masse, en demande beaucoup.
- **Étapes** : construire de nombreux arbres avec des coupures aléatoires (variable et seuil tirés au hasard) → mesurer pour chaque point la profondeur moyenne à laquelle il est isolé → en déduire un score d'anomalie.
- **Formule** : $\displaystyle s(x)=2^{-\,\mathbb E[h(x)]/c(n)}$ ; $h(x)$ = profondeur d'isolement de $x$ dans un arbre, $\mathbb E[h(x)]$ = sa moyenne sur la forêt, $c(n)$ = profondeur moyenne attendue d'un arbre sur $n$ points (normalisation) ; $s$ proche de 1 → anomalie, proche de 0,5 ou moins → normal.
- **Paramètres / hyperparamètres** : paramètres = arbres aléatoires ; hyperparamètres = `n_estimators`, `max_samples`, `contamination` (0,02 ici : part supposée d'anomalies, fixe le seuil).
- **Avantage 1** : rapide (complexité quasi linéaire), sans hypothèse de distribution.
- **Avantage 2** : adapté à de grandes données de dimension élevée.
- **Limite 1** : la proportion `contamination` est une hypothèse qui force le nombre d'alertes.
- **Limite 2** : coupures parallèles aux axes et difficulté avec les anomalies locales proches d'un groupe dense.

### Schéma explicatif

**Comment lire ce schéma.** Le fond coloré est le score appris : bleu au centre (zone dense, normale), rouge sur les bords (zones où un point serait facile à isoler). Les points cerclés de rouge sont les alertes : ce sont les plus éloignés de la masse et ceux dont le score est le plus bas.

```python
def schema():
    from sklearn.ensemble import IsolationForest
    g = np.random.default_rng(16); X = np.r_[g.normal(0, 1, (200, 2)), [[5, 5], [-5.5, 4], [4.5, -5], [-6, -5]]]
    m = IsolationForest(contamination=.02, random_state=0).fit(X); s = m.score_samples(X)
    gx, gy = np.meshgrid(np.linspace(-8, 8, 200), np.linspace(-8, 8, 200)); z = m.score_samples(np.c_[gx.ravel(), gy.ravel()]).reshape(gx.shape)
    fig, ax = plt.subplots(figsize=(6.4, 5)); cs = ax.contourf(gx, gy, z, 15, cmap='RdYlBu'); fig.colorbar(cs, label='score de normalité (bas = atypique)')
    ax.scatter(X[:, 0], X[:, 1], c='k', s=8, label='observations'); a = m.predict(X) == -1
    ax.scatter(X[a, 0], X[a, 1], s=110, facecolors='none', edgecolors='red', lw=2, label='alertes (prédit −1)')
    ax.set_xlabel('$x_1$'); ax.set_ylabel('$x_2$'); ax.set_title('Isolation Forest : zones denses = normales'); ax.legend(fontsize=8, loc='lower right'); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : détection de transactions bancaires ou de notes de frais suspectes sans étiquettes de fraude.
- **Cas d'usage 2 (logistique)** : repérer des capteurs défaillants ou des délais de livraison aberrants.
- **Données et prérequis** : variables numériques ; estimation raisonnable de la part d'anomalies ; si possible, analyse manuelle des alertes.
- **Cas peu approprié** : anomalies contextuelles/temporelles (un montant normal à une heure anormale) sans variables adaptées.
- **Métriques** : sans étiquettes fiables, qualité des alertes jugée par expertise (revue des cas) ; avec étiquettes : précision, rappel, PR-AUC, ROC-AUC.

### Analyse de l'expérience Python

```python
from sklearn.ensemble import IsolationForest
X_anomal=np.vstack([X_clu,np.array([[9,9],[-8,-8]])])
modele=IsolationForest(contamination=.02,random_state=GRAINE).fit(X_anomal)
print('Observations atypiques:',np.where(modele.predict(X_anomal)==-1)[0])
```

**Commentaire.**

**Entrées** : `X_anomal` = les 240 points de `X_clu` plus deux points ajoutés, `[9, 9]` (indice 240) et `[−8, −8]` (indice 241) ; **hyperparamètres** : `contamination=0.02`, `random_state=42`. **Sortie** : observations atypiques aux indices **[20, 79, 130, 147, 240]**.
**Lecture** : le point ajouté `[9, 9]` (indice 240) est bien détecté. Le point `[−8, −8]` (indice 241) ne l'est pas : il se trouve en fait près d'un des groupes centrés vers $(-6{,}9\,;\,-6{,}9)$, il n'est donc pas atypique. Les quatre autres alertes sont des points de bordure de groupes : avec `contamination=0.02` sur 242 points, l'algorithme est forcé de signaler environ 5 points quoi qu'il arrive.
**Limites** : sans étiquettes, on ne peut pas calculer précision ni rappel ; le nombre d'alertes est imposé par `contamination`, pas découvert par l'algorithme.

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Bandits manchots (*Multi-armed bandits*) — **famille** : apprentissage par renforcement (cas à un seul état) ; **objectif** : choisir répétitivement parmi plusieurs options (bras) à gain inconnu pour maximiser le gain cumulé.
- **Intuition** : chaque bras a une probabilité de gain inconnue ; il faut équilibrer *exploration* (tester) et *exploitation* (jouer le meilleur bras estimé).
- **Étapes** : à chaque tour $t$, choisir un bras selon une stratégie → observer la récompense → mettre à jour l'estimation de ce bras.
- **Formule** : estimation $\hat\mu_a(t)=\dfrac{1}{N_a(t)}\sum_{s\le t}r_s\mathbf 1[a_s=a]$ ; regret cumulé $R(T)=T\mu^*-\sum_{t=1}^T\mu_{a_t}$ ; $N_a$ = nombre de tirages du bras $a$, $\mu^*$ = gain moyen du meilleur bras, $r_s$ = récompense au tour $s$.
- **Paramètres / hyperparamètres** : paramètres = estimations $\hat\mu_a$ ; hyperparamètres = règle de sélection (ici un tirage cyclique $t \bmod 3$, soit exploration pure), horizon $T$ (500).
- **Avantage 1** : modèle simple pour décider sous incertitude, sans modèle préalable de l'environnement.
- **Avantage 2** : apprentissage en ligne : on optimise pendant qu'on collecte les données.
- **Limite 1** : la stratégie cyclique de l'exemple ne profite jamais de ce qu'elle apprend (regret linéaire).
- **Limite 2** : le modèle de base ignore le contexte (état du client, saison) et suppose des gains stationnaires.

### Schéma explicatif

**Comment lire ce schéma.** Chaque couleur est un bras. La courbe pleine est la moyenne des gains observés au fil des tirages de ce bras ; le trait pointillé est sa vraie probabilité (inconnue de l'agent). Au début les estimations sont très bruitées, puis elles se rapprochent de la vérité (avec un écart résiduel tant que le nombre de tirages reste faible). (Simulation illustrative avec nos propres graines, non reprise du notebook.)

```python
def schema():
    g = np.random.default_rng(17); vrai = np.array([.2, .5, .7]); S = np.zeros(3); N = np.zeros(3); hist = [[], [], []]
    for t in range(300):
        a = t % 3; r = g.random() < vrai[a]; S[a] += r; N[a] += 1; hist[a].append(S[a]/N[a])
    fig, ax = plt.subplots(figsize=(6.8, 4)); coul = ['#2980b9', '#e67e22', '#27ae60']
    for a in range(3):
        ax.plot(np.arange(1, len(hist[a]) + 1), hist[a], c=coul[a], label=f'bras {a+1} : estimation $\\hat\\mu_{a+1}$'); ax.axhline(vrai[a], c=coul[a], ls='--', lw=1)
    ax.set_ylim(0, 1); ax.set_xlabel('nombre de tirages du bras'); ax.set_ylabel('gain moyen estimé'); ax.set_title('Bandits : les estimations se rapprochent des vraies probabilités (pointillés)'); ax.legend(fontsize=8); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (marketing)** : test A/B continu de plusieurs visuels ou messages publicitaires, en dirigeant progressivement le trafic vers le meilleur.
- **Cas d'usage 2 (finance)** : allocation entre plusieurs stratégies de trading ou canaux de placement à rendement inconnu.
- **Données et prérequis** : retour d'information immédiat après chaque choix, gains approximativement stationnaires.
- **Cas peu approprié** : décisions dont la récompense est très différée ou qui changent l'état du système (il faut alors du renforcement complet).
- **Métriques** : regret cumulé, récompense cumulée, part de tirages du meilleur bras, variation entre graines.

### Analyse de l'expérience Python

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3)
for t in range(500):
    bras=t%3
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Valeurs estimées:',somme/comptes)
```

**Commentaire.**

**Entrées** : trois bras de vraies probabilités $[0{,}2\,;0{,}5\,;0{,}7]$ (simulées, `default_rng(42)`), 500 tirages en rotation (bras $t\bmod3$) ; **sortie** : valeurs estimées **[0,180 ; 0,539 ; 0,723]**.
**Lecture** : chaque bras a été tiré environ 167 fois ; les estimations sont proches des vraies valeurs (écarts de 0,02 à 0,04) et classent correctement les bras.
**Limites** : la stratégie est de l'exploration pure : 2/3 des tirages vont à des bras moins bons, donc le regret est élevé ; la sortie n'affiche ni regret ni récompense cumulée. Les probabilités sont simulées, pas des données réelles.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Epsilon-greedy ($\varepsilon$-glouton) — **famille** : renforcement, stratégie pour les bandits ; **objectif** : équilibrer exploration et exploitation par une règle très simple.
- **Intuition** : la plupart du temps on joue le bras qui semble le meilleur ; avec une petite probabilité $\varepsilon$ on en essaie un au hasard pour continuer à apprendre.
- **Étapes** : à chaque tour, tirer un nombre $u\in[0,1]$ → si $u<\varepsilon$ choisir un bras au hasard, sinon le bras de plus forte estimation → observer la récompense → mettre à jour l'estimation.
- **Formule** : $a_t=\begin{cases}\text{bras aléatoire} & \text{avec prob. }\varepsilon\\ \arg\max_a\hat\mu_a & \text{avec prob. }1-\varepsilon\end{cases}$ ; $\hat\mu_a$ = gain moyen estimé du bras $a$.
- **Paramètres / hyperparamètres** : paramètres = estimations $\hat\mu_a$ ; hyperparamètres = $\varepsilon$ (0,1 ici), horizon (1000), initialisation optimiste des estimations (ici `(somme+1)/(comptes+1)`).
- **Avantage 1** : très simple à coder et à expliquer.
- **Avantage 2** : converge vers le meilleur bras avec un regret bien plus faible qu'une exploration pure.
- **Limite 1** : l'exploration reste aveugle (on essaie aussi des bras clairement mauvais) et un $\varepsilon$ constant garde un regret linéaire.
- **Limite 2** : sensible au choix de $\varepsilon$ ; moins efficace que UCB ou l'échantillonnage de Thompson.

### Schéma explicatif

**Comment lire ce schéma.** À gauche : le diagramme de décision d'un tour. À droite : récompense cumulée moyenne sur 20 simulations pour trois valeurs de $\varepsilon$ ; plus la courbe est proche du pointillé (idéal), plus la stratégie est efficace. $\varepsilon=0$ risque de se bloquer sur un mauvais bras, $\varepsilon$ trop grand gaspille des tirages en exploration. (Simulation illustrative propre à ce schéma.)

```python
def schema():
    vrai = np.array([.2, .5, .7]); T = 1000
    def sim(eps, graine):
        g = np.random.default_rng(graine); S = np.zeros(3); N = np.zeros(3); cum = []; tot = 0
        for t in range(T):
            a = g.integers(3) if g.random() < eps else int(np.argmax(S/np.maximum(N, 1))); r = g.random() < vrai[a]; S[a] += r; N[a] += 1; tot += r; cum.append(tot)
        return np.array(cum)
    fig, ax = plt.subplots(1, 2, figsize=(12.5, 4.2), gridspec_kw={'width_ratios': [1, 1.2]}); toile(ax[0], (0, 10), (0, 6), 'Règle de décision à chaque tour')
    boite(ax[0], 5, 5, 'tirer $u\\sim U(0,1)$', 3.4, .9, '#eaf2f8'); boite(ax[0], 2.3, 2.6, 'EXPLORER\nbras au hasard', 3.2, 1.2, '#fdebd0'); boite(ax[0], 7.7, 2.6, 'EXPLOITER\n$\\arg\\max_a\\hat\\mu_a$', 3.2, 1.2, '#d5f5e3')
    fleche(ax[0], 4, 4.5, 2.7, 3.3, '$u<\\varepsilon$', decalage=.3); fleche(ax[0], 6, 4.5, 7.3, 3.3, '$u\\geq\\varepsilon$', decalage=.3); boite(ax[0], 5, .7, 'récompense → mise à jour de $\\hat\\mu_a$', 5.5, .9, '#fcf3cf')
    fleche(ax[0], 2.3, 1.95, 3.6, 1.1); fleche(ax[0], 7.7, 1.95, 6.4, 1.1)
    for eps, c in [(0, '#c0392b'), (.1, '#27ae60'), (.5, '#2980b9')]:
        ax[1].plot(np.mean([sim(eps, s) for s in range(20)], axis=0), c=c, label=rf'$\varepsilon$={eps}')
    ax[1].plot(np.arange(1, T + 1)*vrai.max(), 'k--', lw=1, label='idéal (toujours le meilleur bras)'); ax[1].set_xlabel('tour'); ax[1].set_ylabel('récompense cumulée moyenne (20 graines)'); ax[1].set_title('Effet de $\\varepsilon$'); ax[1].legend(fontsize=8); plt.tight_layout(); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (marketing)** : choisir entre plusieurs bannières ou promotions en testant 10 % du temps les alternatives.
- **Cas d'usage 2 (finance / gestion)** : orienter des ordres vers le courtier ou la plateforme d'exécution au meilleur coût observé.
- **Données et prérequis** : retour de récompense rapide, bras peu nombreux, environnement stable.
- **Cas peu approprié** : environnement fortement non stationnaire (sans oubli des anciennes données) ou erreurs coûteuses (santé, risque élevé) où explorer au hasard est inacceptable.
- **Métriques** : récompense cumulée, regret, part de tirages du meilleur bras, variation entre graines.

### Analyse de l'expérience Python

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

**Commentaire.**

**Entrées** : mêmes trois bras simulés ; **hyperparamètres** : `epsilon=0.1`, 1000 tirages, estimation `(somme+1)/(comptes+1)` (initialisation optimiste). **Sortie** : nombre de choix = **[25, 49, 926]**, valeurs estimées = **[0,200 ; 0,673 ; 0,693]**.
**Lecture** : 92,6 % des tirages vont au meilleur bras (bras 3), contre un tiers avec la stratégie cyclique. Le bras 2 est estimé à 0,673 alors que sa vraie valeur est 0,5 : il n'a été tiré que 49 fois, l'estimation est donc encore bruitée (et gonflée par l'initialisation optimiste).
**Limites** : une seule graine ; on ne mesure ni le regret ni la variation entre graines, et le même résultat ne suffit pas à choisir $\varepsilon=0{,}1$ comme optimal.

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : Q-learning — **famille** : apprentissage par renforcement (valeur, hors politique, tabulaire) ; **objectif** : apprendre une politique optimale en estimant la valeur de chaque couple (état, action).
- **Intuition** : l'agent apprend un tableau $Q(s,a)$ = « récompense future totale attendue si je fais $a$ dans $s$ puis agis au mieux ».
- **Étapes** : observer $s$ → choisir $a$ ($\varepsilon$-glouton) → observer $r$ et $s'$ → corriger $Q(s,a)$ vers la cible → répéter sur de nombreux épisodes.
- **Formule** : $\displaystyle Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\big]$ ; $\alpha$ = taux d'apprentissage (0,3), $\gamma$ = facteur d'actualisation (0,95), $r$ = récompense, $s'$ = état suivant, $a'$ = action possible en $s'$.
- **Paramètres / hyperparamètres** : paramètres = table $Q$ ; hyperparamètres = $\alpha$, $\gamma$, $\varepsilon$ (décroissant de 1 à 0,05), nombre d'épisodes (4000).
- **Avantage 1** : simple et converge vers la politique optimale (sous conditions d'exploration).
- **Avantage 2** : hors politique : apprend la politique optimale même en explorant.
- **Limite 1** : table inutilisable avec trop d'états (nécessite des approximations, ex. DQN).
- **Limite 2** : peut être instable/optimiste (surestimation due au `max`) et sensible aux hyperparamètres.

### Schéma explicatif

**Comment lire ce schéma.** Le cycle se lit en boucle : l'agent (à gauche) envoie une action à l'environnement (à droite) ; celui-ci renvoie le nouvel état et une récompense. L'encadré jaune est la règle de mise à jour du tableau $Q$ : on rapproche l'ancienne valeur d'une cible qui combine la récompense obtenue et la meilleure valeur estimée à l'état suivant.

```python
def schema():
    fig, ax = plt.subplots(figsize=(10, 4.4)); boucle_agent(ax, r'$Q(s_t,a_t)\leftarrow Q(s_t,a_t)+\alpha\,[\,r_{t+1}+\gamma\,\max_{a\prime}Q(s_{t+1},a\prime)-Q(s_t,a_t)\,]$', 'hors politique : la cible utilise le MEILLEUR coup suivant', 'Q-learning : boucle agent–environnement'); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (logistique)** : apprendre un itinéraire ou une politique de gestion de stock avec peu d'états discrets.
- **Cas d'usage 2 (finance)** : politique d'exécution simplifiée d'un ordre (acheter/attendre) dans un environnement simulé discrétisé.
- **Données et prérequis** : environnement simulable avec états et actions discrets et peu nombreux (Gymnasium) ; nombreuses interactions.
- **Cas peu approprié** : espaces d'états continus ou immenses ; apprentissage sur le réel sans simulateur (explorer coûte cher).
- **Métriques** : récompense cumulée par épisode, taux de réussite, vitesse de convergence, variation entre graines.

### Analyse de l'expérience Python

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

**Commentaire.**

**Entrées** : environnement Gymnasium `FrozenLake-v1` 4×4 **sans glissement** (`is_slippery=False`, déterministe), 16 états × 4 actions ; **hyperparamètres** : $\alpha=0{,}3$, $\gamma=0{,}95$, $\varepsilon$ décroissant de 1 à 0,05, 4000 épisodes. **Sortie** : un vecteur de 16 actions (la politique gloutonne `Q.argmax(axis=1)`), codées 0 = gauche, 1 = bas, 2 = droite, 3 = haut.
**Résultat** : cette cellule nécessite `gymnasium`, installé par la première cellule ; je n'ai pas pu l'exécuter pour reporter votre vecteur de politique, il faut le lire dans votre exécution. À vérifier : en suivant les actions depuis la case de départ, atteint-on l'arrivée sans tomber dans un trou ?
**Ce que la sortie permet / ne permet pas** : elle donne une politique, mais pas de récompense cumulée ni de taux de réussite ; pour conclure que l'agent a appris, il faut rejouer la politique. L'environnement étant déterministe, le résultat est plus facile qu'avec glissement.

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : SARSA (*State–Action–Reward–State–Action*) — **famille** : apprentissage par renforcement (valeur, sur politique, tabulaire) ; **objectif** : apprendre la valeur de la politique effectivement suivie, exploration comprise.
- **Intuition** : proche du Q-learning, mais la cible utilise l'action $a'$ réellement choisie à l'étape suivante, pas la meilleure action possible : l'agent tient compte de ses propres exploration et erreurs.
- **Étapes** : observer $s$, choisir $a$ → recevoir $r$, $s'$ → choisir $a'$ ($\varepsilon$-glouton) → mettre à jour $Q(s,a)$ avec $Q(s',a')$ → avancer ($s\leftarrow s'$, $a\leftarrow a'$).
- **Formule** : $\displaystyle Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\,Q(s',a')-Q(s,a)\big]$ ; $a'$ = action effectivement choisie en $s'$ ; les autres symboles sont ceux du Q-learning ($\alpha$ = 0,3, $\gamma$ = 0,95).
- **Paramètres / hyperparamètres** : paramètres = table $Q$ ; hyperparamètres = $\alpha$, $\gamma$, schéma de décroissance de $\varepsilon$, épisodes (4000).
- **Avantage 1** : apprend des politiques plus prudentes lorsque l'exploration est risquée (ex. falaise).
- **Avantage 2** : converge vers la politique optimale si $\varepsilon\to0$ progressivement.
- **Limite 1** : sur politique : moins efficace en données que le Q-learning pour atteindre la politique optimale.
- **Limite 2** : même limite de passage à l'échelle que toutes les méthodes tabulaires.

### Schéma explicatif

**Comment lire ce schéma.** À gauche, l'état suivant $s'$ ; à droite, les valeurs estimées de ses quatre actions (valeurs illustratives). Le Q-learning (vert) utilise toujours la meilleure valeur (2,5). SARSA (orange) utilise la valeur de l'action que l'agent a réellement choisie — ici une action d'exploration (1,8) — donc sa cible intègre le risque de ses propres choix exploratoires.

```python
def schema():
    fig, ax = plt.subplots(figsize=(10, 4.6)); toile(ax, (0, 10), (0, 5.4), 'Cible de mise à jour dans l\'état suivant $s\'$ (valeurs $Q$ illustratives)')
    boite(ax, 1.4, 2.7, 'état $s\'$', 1.8, 1, '#eaf2f8'); vals = [1.0, 2.5, .4, 1.8]; ys = [4.6, 3.3, 2.0, .7]
    for k, (v, y) in enumerate(zip(vals, ys)):
        coul = '#d5f5e3' if k == 1 else '#fdebd0' if k == 3 else '#f2f3f4'; boite(ax, 5, y, f'$Q(s\',a_{k+1})={v}$', 2.8, .8, coul); fleche(ax, 2.35, 2.7, 3.55, y)
    ax.text(7.2, 3.3, '← Q-learning : prend le MAX\n   cible = r + γ·2,5', fontsize=9, va='center', color='#1e8449'); ax.text(7.2, .7, '← SARSA : prend l\'action $a\'$ réellement\n   choisie (ε-glouton), ici $a_4$\n   cible = r + γ·1,8', fontsize=9, va='center', color='#af601a'); plt.show()
schema()
```

### Contextes d'utilisation

- **Cas d'usage 1 (logistique)** : planification de trajectoire d'un robot ou véhicule où un pas d'exploration peut être coûteux (« chemin sûr »).
- **Cas d'usage 2 (finance / gestion)** : politique de trading simulée où l'on veut évaluer le comportement réel de la stratégie, exploration comprise.
- **Données et prérequis** : environnement simulable à états et actions discrets ; $\varepsilon$ décroissant.
- **Cas peu approprié** : si l'on veut seulement la politique optimale gloutonne sans se soucier de l'exploration (Q-learning), ou espaces continus.
- **Métriques** : récompense cumulée *pendant* l'apprentissage et à l'évaluation, taux de réussite, variation entre graines.

### Analyse de l'expérience Python

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

**Commentaire.**

**Entrées** : même environnement `FrozenLake-v1` déterministe ; mêmes hyperparamètres que pour le Q-learning ; l'action suivante est choisie *avant* la mise à jour. **Sortie** : le vecteur de 16 actions gloutonnes.
**Résultat** : cellule dépendant de `gymnasium`, que je n'ai pas pu exécuter ici ; relevez le vecteur dans votre exécution et comparez-le à celui du Q-learning (cellule 27).
**Ce que la sortie permet / ne permet pas** : dans un environnement déterministe sans falaise, Q-learning et SARSA peuvent donner des politiques proches voire identiques ; la différence de prudence que SARSA est censé produire ne se voit pas vraiment ici (elle apparaît dans un environnement comme *CliffWalking*). ## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom** : DQN (*Deep Q-Network*) — **famille** : apprentissage par renforcement profond (valeur) ; **objectif** : approximer la fonction $Q(s,a)$ par un réseau de neurones pour traiter de grands (ou continus) espaces d'états.
- **Intuition** : au lieu d'une table, un réseau prend l'état en entrée et sort une valeur par action ; on l'entraîne à imiter la cible de Bellman.
- **Étapes** : agir ($\varepsilon$-glouton) → stocker $(s,a,r,s')$ dans une mémoire de rejeu → tirer un mini-lot au hasard → calculer la cible avec un réseau cible figé → descendre le gradient sur le réseau en ligne → copier périodiquement ses poids dans le réseau cible.
- **Formule** : cible $y=r+\gamma(1-d)\max_{a'}Q_{\theta^-}(s',a')$ et perte $L(\theta)=\mathbb E\big[(y-Q_\theta(s,a))^2\big]$ ; $\theta$ = poids du réseau en ligne, $\theta^-$ = poids du réseau cible, $d\in\{0,1\}$ = fin d'épisode.
- **Paramètres / hyperparamètres** : paramètres = poids $\theta$ ; hyperparamètres = $\gamma$ (0,95 dans l'exemple), taille de la mémoire et du lot, fréquence de copie, pas d'apprentissage, $\varepsilon$.
- **Avantage 1** : traite des états continus/de grande dimension (images, signaux) sans table.
- **Avantage 2** : la mémoire de rejeu et le réseau cible stabilisent l'apprentissage.
- **Limite 1** : peu efficace en échantillons, sensible aux hyperparamètres et aux graines (instable).
- **Limite 2** : coûteux (réseau, GPU) et difficile à interpréter ; surestimation des valeurs.

### Schéma explicatif

**Comment lire ce schéma.** Le schéma se lit de gauche à droite. L'environnement produit des transitions rangées dans la mémoire de rejeu ; des mini-lots sont tirés au hasard. Le réseau cible (figé) fournit la cible $y$, le réseau en ligne fournit la prédiction ; leur écart définit la perte dont le gradient met à jour le réseau en ligne. Toutes les $C$ étapes, ses poids sont copiés dans le réseau cible. La flèche du bas montre que le réseau en ligne choisit aussi les actions.

### Contextes d'utilisation

- **Cas d'usage 1 (finance)** : exécution optimale d'ordres ou gestion de portefeuille en simulation, avec un état de marché de grande dimension.
- **Cas d'usage 2 (logistique)** : pilotage de gestion de stocks ou de flottes avec de nombreux états continus.
- **Données et prérequis** : simulateur (Gymnasium ou environnement métier), beaucoup d'interactions, GPU conseillé, bibliothèque de réseaux (PyTorch/TensorFlow).
- **Cas peu approprié** : petits espaces d'états (une table suffit), absence de simulateur, ou besoin de garanties de sécurité/explicabilité.
- **Métriques** : récompense cumulée par épisode, taux de réussite, stabilité entre graines, perte de Bellman (indicative seulement).

### Analyse de l'expérience Python

```python
# Calcul de cible DQN, sans entraîner un réseau neuronal complet
recompenses=np.array([0.,1.,-1.]);termine=np.array([False,True,False]);meilleur_Q_suivant=np.array([2.,3.,1.]);gamma=.95
cibles=recompenses+gamma*(~termine)*meilleur_Q_suivant
print('Cibles de Bellman:',cibles)
# Un DQN complet ajoute réseau en ligne, réseau cible et mémoire de rejeu.
```

**Commentaire.**

**Entrées** : trois transitions fictives avec récompenses `[0, 1, −1]`, indicateurs de fin d'épisode `[faux, vrai, faux]`, meilleures valeurs suivantes `[2, 3, 1]` et $\gamma=0{,}95$ ; **sortie** : cibles de Bellman = **[1,9 ; 1,0 ; −0,05]**.
**Lecture** : cible $=r+\gamma(1-d)\max Q$ : $0+0{,}95\times2=1{,}9$ ; $1+0=1$ (épisode terminé, la valeur future est ignorée) ; $-1+0{,}95\times1=-0{,}05$.
**Limites** : cette cellule est **conceptuelle** : aucun réseau n'est entraîné, il n'y a ni mémoire de rejeu ni réseau cible ; les valeurs sont inventées pour la démonstration. Une expérience DQN complète est facultative et n'est pas traitée ici.
